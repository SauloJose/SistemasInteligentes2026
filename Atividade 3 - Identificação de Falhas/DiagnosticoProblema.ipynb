## 2. Unidades de Processo

| Unidade | Função principal |
|---------|------------------|
| **Reator** | Ocorrem as reações exotérmicas entre os reagentes A, C, D, E, gerando produtos G, H e subproduto F. Possui agitação e sistema de resfriamento. |

| **Corrente de purga** | Purga uma fração do gás reciclado para evitar acúmulo de inertes (B) e subproduto, controlada por uma válvula. |

# Testes $T^2$ e $Q$ para detecção de anomalias com PLS

Quando se utiliza um modelo **PLS** (Mínimos Quadrados Parciais) para monitorar um processo com múltiplas entradas ($\mathbf{X}$) e múltiplas saídas ($\mathbf{Y}$), as estatísticas **$T^2$ de Hotelling** e **$Q$ (SPE – Squared Prediction Error)** permitem detectar desvios em relação ao comportamento normal. Ambas são calculadas a partir dos escores do modelo construído com dados de referência.



## 1. Modelo PLS (breve revisão)

Para $N$ observações normais, o PLS decompõe as matrizes da seguinte forma:

$$\mathbf{X} = \mathbf{T} \mathbf{P}^\top + \mathbf{E}$$

$$\mathbf{Y} = \mathbf{T} \mathbf{Q}^\top + \mathbf{F}$$

* **$\mathbf{T}$** ($N \times A$): Matriz de escores (scores), ortogonais entre si.
* **$\mathbf{P}$** e **$\mathbf{Q}$**: Loadings de $\mathbf{X}$ e $\mathbf{Y}$, respectivamente.
* **$A$**: Número de componentes latentes selecionadas.

Para uma **nova amostra** $(\mathbf{x}_{\text{new}}, \mathbf{y}_{\text{new}})$, obtemos os escores projetados:

$$\mathbf{t}_{\text{new}} = \mathbf{x}_{\text{new}}^\top \mathbf{R}$$


onde $\mathbf{R} = \mathbf{W}(\mathbf{P}^\top \mathbf{W})^{-1}$ representa a matriz de pesos de projeção.



## 2. Estatística $T^2$ (Hotelling)

### O que mede

A **distância dos escores** da nova observação ao centro dos escores de treino, dentro do subespaço latente. Detecta variações **sistemáticas** (mudanças no ponto de operação) que ainda respeitam a estrutura de correlação do modelo.

### Fórmula

$$T^2 = \mathbf{t}_{\text{new}}^\top \mathbf{S}^{-1} \mathbf{t}_{\text{new}} = \sum_{a=1}^{A} \frac{t_{\text{new},a}^2}{\lambda_a}$$

* $\mathbf{S} = \frac{\mathbf{T}_{\text{treino}}^\top \mathbf{T}_{\text{treino}}}{N-1}$ (matriz de covariância dos escores).
* $\lambda_a = \text{Var}(t_a)$ no conjunto de treinamento.

### Limite de controle

$$T^2_{\lim} = \frac{A\,(N^2 - 1)}{N\,(N - A)}\; F_{A,\, N-A,\; \alpha}$$


onde $F$ é o quantil da distribuição de Fisher com nível de significância $\alpha$ (ex: 0.95 ou 0.99).

> **Interpretação:** $T^2$ alto $\rightarrow$ A amostra está longe da média, mas o modelo ainda "entende" a relação entre as variáveis.


## 3. Estatística $Q$ (SPE – Squared Prediction Error)

### O que mede

A **falta de ajuste** do modelo, ou seja, a variabilidade **não explicada** pelas componentes latentes. Detecta quebras na estrutura de correlação, como falhas de sensores ou mudanças súbitas na dinâmica do sistema.

### Fórmulas de Resíduos

**Resíduos de $\mathbf{X}$:**


$$\mathbf{e}_x = \mathbf{x}_{\text{new}} - \mathbf{P}\,\mathbf{t}_{\text{new}} \implies Q_x = \mathbf{e}_x^\top \mathbf{e}_x = \|\mathbf{e}_x\|^2$$

**Resíduos de $\mathbf{Y}$:**


$$\mathbf{e}_y = \mathbf{y}_{\text{new}} - \mathbf{Q}\,\mathbf{t}_{\text{new}} \implies Q_y = \mathbf{e}_y^\top \mathbf{e}_y = \|\mathbf{e}_y\|^2$$

### Limite de controle (Aproximação de Box)

Calculado a partir da média ($m$) e variância ($v$) dos valores de $Q$ no treino:


$$g = \frac{v}{2m}, \qquad h = \frac{2m^2}{v} \implies Q_{\lim} = g \cdot \chi^2_{h,\; \alpha}$$



## 4. Implementação Prática e Diagnóstico

### 🔹 Estratégia de Monitoramento

Para cada nova amostra, comparamos os índices com os limites calculados na fase de treino:

1. **$T^2 \uparrow$, $Q$ normal:** O processo mudou de regime (ex: aumento de produção), mas os sensores continuam correlacionados.
2. **$Q \uparrow$, $T^2$ normal:** Falha técnica ou anomalia estrutural (ex: sensor travado ou vazamento).
3. **Ambos $\uparrow$:** Anomalia grave e fora do espaço de operação conhecido.

# Implementação do Algorítmo

## Primeira Etapa: Extrair dados desejados para análise

````python
from mpl_toolkits.mplot3d import Axes3D
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cross_decomposition import PLSRegression
from sklearn.preprocessing import StandardScaler
from scipy.stats import chi2
from IPython.display import display, Markdown
from numba import njit 
#import simpy as sp

# Importando todas as funções do filtro de Kalman
from utils import *

# Configuração de estilo para os gráficos ficarem mais legíveis
plt.style.use('seaborn-v0_8-whitegrid')
````

````python
#Nome das Colunas
colums = [f"XML{i}" for i in range(1,53)]

# Puxar uma série temporal com pandas para testar
df = pd.read_csv('TE_process/d05_te.dat', sep='\s+', header=None,names=colums)

# Exibindo o elemento correspondente da tabela
display(df)

#DEFININDO O NOME DAS VARIÁVEIS QUE SERÃO TESTADAS
x_name = "XML7"
y_name = "XML7"

# Puxando a Série temporal
serie = np.array(df[x_name], dtype=np.float64) #960 colunas

# Printando a dimensão da série 
print(serie.shape)

#Dimensões
N = len(serie)

#Plotar o gráfico da série
plt.plot(serie)
````

## Segunda Etapa: Definição dos dados de treino

````python
# Separando X e Y para o treino
X_train = df[[x_name]]
Y_train = df[[y_name]]

# Padronização (Z-score) é obrigatória para PLS
scaler_X = StandardScaler()
scaler_Y = StandardScaler()

X_train_sc = scaler_X.fit_transform(X_train)
Y_train_sc = scaler_Y.fit_transform(Y_train)

# =====================================================================
# 2. Treinar um PLS em cima do resultado normal
# =====================================================================
# Como temos 1 variável X predizendo 1 Y, usaremos 1 componente latente
n_components = 1 
pls = PLSRegression(n_components=n_components)
pls.fit(X_train_sc, Y_train_sc)
````

## Terceira Etapa: Definindo limites de confiança para o PLS nos testes T² e Q.


````python
# =====================================================================
# 3. Implementar código do teste T² e Q
# =====================================================================
# Extraindo scores (T) do modelo para os dados de treino
T_train = pls.x_scores_

# Matriz de covariância dos scores (usada no T²)
cov_T = np.cov(T_train.T)
if n_components == 1:
    cov_T = np.array([[cov_T]]) # Garantir que seja 2D

# Função genérica para calcular T² e Q para qualquer conjunto de dados
def calcular_metricas(X, Y, pls_model, cov_T):
    # Calcula os scores T para a nova amostra
    T = pls_model.transform(X)
    
    # Cálculo do T² (Hotelling)
    T2 = np.zeros(X.shape[0])
    inv_cov = np.linalg.inv(cov_T)
    for i in range(X.shape[0]):
        T2[i] = T[i, :].T @ inv_cov @ T[i, :]
        
    # Cálculo do Q (Resíduos da predição em Y)
    Y_pred = pls_model.predict(X)
    E = Y - Y_pred
    Q = np.sum(E**2, axis=1) # Erro de predição quadrático
    
    return T2, Q

T2_train, Q_train = calcular_metricas(X_train_sc, Y_train_sc, pls, cov_T)
````

````python
# =====================================================================
# 4. Definir limites de aplicação para T² e Q
# =====================================================================
confianca = 0.95 # Limite de 95% de confiança estatística

# Limite para T² (Aproximação pela distribuição Qui-quadrado)
T2_limite = chi2.ppf(confianca, df=n_components)

# Limite para Q (Aproximação empírica pelo percentil de 95% dos dados normais)
# Nota: Para dados unidimensionais, o percentil é mais robusto que a aproximação de Jackson-Mudholkar
Q_limite = np.percentile(Q_train, confianca * 100)

print(f"--- Limites de Controle Definidos ---")
print(f"Limite T² (95%): {T2_limite:.4f}")
print(f"Limite Q  (95%): {Q_limite:.4f}")
````

## Quarta Etapa: Detecção de Anomalias encontradas e interpretação dos dados

````python
# =====================================================================
# 5. Fazer testes com os valores do banco de dados e Anotar
# =====================================================================
# Preparando os novos dados
X_test_sc = scaler_X.transform(df[[x_name]])
Y_test_sc = scaler_Y.transform(df[[y_name]])

# Calculando as métricas para os novos dados
T2_test, Q_test = calcular_metricas(X_test_sc, Y_test_sc, pls, cov_T)

# Anotando os resultados: Criando DataFrames com os diagnósticos
df_resultados = pd.DataFrame({
    'T2': T2_test,
    'Q': Q_test,
    'Alarme_T2': T2_test > T2_limite,
    'Alarme_Q': Q_test > Q_limite
})

# Extrair as amostras problemáticas
anomalias = df_resultados[(df_resultados['Alarme_T2'] == True) | (df_resultados['Alarme_Q'] == True)]
print(f"\nForam encontradas {len(anomalias)} anomalias em um total de {len(df)} amostras.")
````

````python
# =====================================================================
# VISUALIZAÇÃO GRÁFICA DOS RESULTADOS
# =====================================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

# Gráfico do T²
ax1.plot(df_resultados.index, df_resultados['T2'], marker='o', linestyle='-', color='blue', label='Estatística T²')
ax1.axhline(y=T2_limite, color='red', linestyle='--', label=f'Limite de {confianca*100}%')
ax1.set_title('Gráfico de Controle T² (Desvios na Variável X)')
ax1.set_ylabel('T²')
ax1.legend()
ax1.grid(True)

# Gráfico do Q
ax2.plot(df_resultados.index, df_resultados['Q'], marker='s', linestyle='-', color='green', label='Estatística Q')
ax2.axhline(y=Q_limite, color='red', linestyle='--', label=f'Limite de {confianca*100}%')
ax2.set_title('Gráfico de Controle Q (Quebra de Correlação entre X e Y)')
ax2.set_ylabel('Q (Erro Quadrático)')
ax2.set_xlabel('Índice da Amostra')
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()

# Exportando anotações para análise profunda
# df_resultados.to_csv('resultados_auditoria_pls.csv', index=False)
````

Agora o intuito é recriar esse processo em forma de um Wrapper para detecção das inconveniências

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cross_decomposition import PLSRegression
from sklearn.preprocessing import StandardScaler
from scipy.stats import chi2
import warnings

warnings.filterwarnings("ignore")

# =====================================================================
# 1. FUNÇÕES AUXILIARES
# =====================================================================

def preparar_dados(df, colunas_x, colunas_y):
    X = df[colunas_x]
    Y = df[colunas_y]
    scaler_X = StandardScaler()
    scaler_Y = StandardScaler()
    return scaler_X.fit_transform(X), scaler_Y.fit_transform(Y), scaler_X, scaler_Y

def treinar_modelo_pls(X_train_sc, Y_train_sc, n_components):
    pls = PLSRegression(n_components=n_components)
    pls.fit(X_train_sc, Y_train_sc)
    cov_T = np.cov(pls.x_scores_.T)
    if n_components == 1:
        cov_T = np.array([[cov_T]])
    return pls, cov_T

def calcular_metricas_completas(X_sc, Y_sc, pls_model, cov_T):
    """Calcula T2, Q_X (para as entradas) e Q_Y (para a saída)."""
    # 1. Scores (T) e Hotelling T2
    T = pls_model.transform(X_sc)
    T2 = np.zeros(X_sc.shape[0])
    inv_cov = np.linalg.inv(cov_T)
    for i in range(X_sc.shape[0]):
        T2[i] = T[i, :].T @ inv_cov @ T[i, :]
        
    # 2. Reconstrução de X e Q_X (Resíduos de Entrada)
    # X_pred = T * P^T
    X_pred = T @ pls_model.x_loadings_.T
    E_x = X_sc - X_pred
    Q_x = np.sum(E_x**2, axis=1)
    
    # 3. Predição de Y e Q_Y (Resíduos de Saída)
    Y_pred = pls_model.predict(X_sc)
    E_y = Y_sc - Y_pred
    Q_y = np.sum(E_y**2, axis=1)
    
    return T2, Q_x, Q_y, E_x

def plotar_monitoramento(df_res, limites, confianca):
    """Plota T2, Q_X e Q_Y em 3 painéis."""
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 10))

    # T2
    ax1.plot(df_res.index, df_res['T2'], marker='o', color='blue', label='T²')
    ax1.axhline(y=limites['T2'], color='red', linestyle='--', label=f'Limite {confianca*100}%')
    ax1.set_title('Controle T² (Desvios Globais em X)')
    ax1.legend()

    # Q_X
    ax2.plot(df_res.index, df_res['Q_X'], marker='s', color='orange', label='Q_X')
    ax2.axhline(y=limites['Q_X'], color='red', linestyle='--', label=f'Limite {confianca*100}%')
    ax2.set_title('Controle Q_X (Erros/Quebras de Correlação nas Entradas X)')
    ax2.legend()

    # Q_Y
    ax3.plot(df_res.index, df_res['Q_Y'], marker='^', color='green', label='Q_Y')
    ax3.axhline(y=limites['Q_Y'], color='red', linestyle='--', label=f'Limite {confianca*100}%')
    ax3.set_title('Controle Q_Y (Erro de Predição na Qualidade Y)')
    ax3.set_xlabel('Índice da Amostra')
    ax3.legend()

    plt.tight_layout()
    plt.show()

def plotar_contribuicao_falha(amostra_idx, E_x, colunas_x):
    """Gera um gráfico de barras para identificar a variável que mais contribuiu para o Q_X."""
    # A contribuição é o erro quadrático individual da variável
    contribuicoes = E_x[amostra_idx] ** 2
    
    plt.figure(figsize=(10, 5))
    plt.bar(colunas_x, contribuicoes, color='crimson')
    plt.title(f'Análise de Contribuição para Q_X (Amostra {amostra_idx})')
    plt.ylabel('Contribuição (Erro Quadrático)')
    plt.xlabel('Variáveis de Entrada (X)')
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--')
    plt.tight_layout()
    plt.show()

def printar_anomalias(resultados):
    total_amostras = len(resultados)
    qtd_t2 = resultados['Alarme_T2'].sum()
    qtd_qx = resultados['Alarme_Q_X'].sum()
    qtd_qy = resultados['Alarme_Q_Y'].sum()
    
    print(f"\n=== RESUMO DE ANOMALIAS (Teste com {total_amostras} amostras) ===")
    print(f"Alarmes T²  (Desvios globais em X)        : {qtd_t2}")
    print(f"Alarmes Q_X (Quebra de correlação em X)   : {qtd_qx}")
    print(f"Alarmes Q_Y (Erro na predição da Qualidade): {qtd_qy}")
    print("==============================================================\n")
# =====================================================================
# 2. WRAPPER PRINCIPAL
# =====================================================================

def analisar_qualidade_pls(df_treino, df_teste, colunas_x, colunas_y, n_components=1, confianca=0.95):
    X_train_sc, Y_train_sc, scaler_X, scaler_Y = preparar_dados(df_treino, colunas_x, colunas_y)
    pls, cov_T = treinar_modelo_pls(X_train_sc, Y_train_sc, n_components)
    
    # Métricas e limites do Treino
    T2_train, Q_x_train, Q_y_train, _ = calcular_metricas_completas(X_train_sc, Y_train_sc, pls, cov_T)
    limites = {
        'T2': chi2.ppf(confianca, df=n_components),
        'Q_X': np.percentile(Q_x_train, confianca * 100),
        'Q_Y': np.percentile(Q_y_train, confianca * 100)
    }
    
    # Teste
    X_test_sc = scaler_X.transform(df_teste[colunas_x])
    Y_test_sc = scaler_Y.transform(df_teste[colunas_y])
    T2_test, Q_x_test, Q_y_test, E_x_test = calcular_metricas_completas(X_test_sc, Y_test_sc, pls, cov_T)
    
    # Tabela de resultados
    df_resultados = pd.DataFrame({
        'T2': T2_test,
        'Q_X': Q_x_test,
        'Q_Y': Q_y_test,
        'Alarme_T2': T2_test > limites['T2'],
        'Alarme_Q_X': Q_x_test > limites['Q_X'],
        'Alarme_Q_Y': Q_y_test > limites['Q_Y']
    })
    
    plotar_monitoramento(df_resultados, limites, confianca)
    
    # Retornamos E_x_test para permitir a análise de contribuição
    return df_resultados, E_x_test
````

## Escolhendo base de dados e dados para analisar

````python
# Simulação da leitura do seu banco de dados
colunas = [f"XML{i}" for i in range(1,53)]
df_completo = pd.read_csv('TE_process/d05_te.dat', sep='\s+', header=None, names=colunas)
    
# Definindo as variáveis
variaveis_entrada = ['XML1', 'XML2', 'XML3', 'XML4'] # Múltiplos Xs
variavel_qualidade = ['XML7']                        # O Y de interesse
    
# Simulação: Separando os primeiros 500 para treino (funcionamento normal) e o resto para teste
df_treino = df_completo.iloc[:500]
df_teste = df_completo.iloc[500:]
````

````python
# Chamando o Wrapper
# 1. Rodar a análise
resultados, erros_x_teste = analisar_qualidade_pls(
        df_treino, df_teste, variaveis_entrada, variavel_qualidade, n_components=2
    )

printar_anomalias(resultados)
    
    # 2. Investigar a primeira anomalia no Q_X que encontrarmos!
anomalias_qx = resultados[resultados['Alarme_Q_X'] == True]
    
if not anomalias_qx.empty:
    primeira_falha_idx = anomalias_qx.index[0]
    print(f"\nAlarme Q_X detectado na amostra {primeira_falha_idx}! Gerando gráfico de contribuição...")
    plotar_contribuicao_falha(primeira_falha_idx, erros_x_teste, variaveis_entrada)
````
