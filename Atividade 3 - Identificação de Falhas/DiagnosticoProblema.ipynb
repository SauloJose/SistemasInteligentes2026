## 2. Unidades de Processo

| Unidade | Função principal |
|---------|------------------|
| **Reator** | Ocorrem as reações exotérmicas entre os reagentes A, C, D, E, gerando produtos G, H e subproduto F. Possui agitação e sistema de resfriamento. |
| **Condensador** | Resfria a corrente de saída do reator, condensando parcialmente os vapores. Utiliza água de resfriamento. |
| **Compressor** | Comprime a fase vapor não condensada e a recicla de volta ao reator, mantendo a circulação de gases. |
| **Separador (Vaso Vapor‑Líquido)** | Separa a mistura bifásica: uma corrente líquida (que segue para o stripper) e uma corrente vapor (reciclada via compressor). |
| **Stripper (Coluna de Esgotamento)** | Remove componentes leves residuais do líquido, produzindo os produtos finais G e H na corrente de fundo. Utiliza vapor como fonte de calor. |
| **Corrente de purga** | Purga uma fração do gás reciclado para evitar acúmulo de inertes (B) e subproduto, controlada por uma válvula. |

# 2. Definição dos Teste de Hipótese que serão utilizados

## 2.1 Estatística $T^2$ (Hotelling)
### 2.2.1. O que mede

A **distância dos escores** da nova observação ao centro dos escores de treino, dentro do subespaço latente. Detecta variações **sistemáticas** (mudanças no ponto de operação) que ainda respeitam a estrutura de correlação do modelo.

### 2.2.2. Fórmula

$$T^2 = \mathbf{t}_{\text{new}}^\top \mathbf{S}^{-1} \mathbf{t}_{\text{new}} = \sum_{a=1}^{A} \frac{t_{\text{new},a}^2}{\lambda_a}$$

* $\mathbf{S} = \frac{\mathbf{T}_{\text{treino}}^\top \mathbf{T}_{\text{treino}}}{N-1}$ (matriz de covariância dos escores).
* $\lambda_a = \text{Var}(t_a)$ no conjunto de treinamento.

### 2.2.3. Limite de controle

$$T^2_{\lim} = \frac{A\,(N^2 - 1)}{N\,(N - A)}\; F_{A,\, N-A,\; \alpha}$$


onde $F$ é o quantil da distribuição de Fisher com nível de significância $\alpha$ (ex: 0.95 ou 0.99).

> **Interpretação:** $T^2$ alto $\rightarrow$ A amostra está longe da média, mas o modelo ainda "entende" a relação entre as variáveis.

## 2.3 Estatística SPE (Squared Prediction Error) - Estatística Q

### 2.3.1. O que mede

A variabilidade **não explicada** pelo modelo. Detecta quebras na estrutura de correlação do processo (ex.: falhas de sensores, mudanças dinâmicas).

### 2.3.2. Formulação
Para uma nova amostra $\mathbf{x}_{\text{new}}$:
- Projeção no espaço principal: $\mathbf{t}_{\text{new}} = \mathbf{x}_{\text{new}} \mathbf{P}$
- Resíduo: $\mathbf{e} = \mathbf{x}_{\text{new}} - \mathbf{t}_{\text{new}} \mathbf{P}^\top$
- **SPE** = $\mathbf{e}\mathbf{e}^\top = \|\mathbf{e}\|^2$

No código:
```python
SPE = z_test[i].T @ P_res @ P_res.T @ z_test[i]
```

### 2.3.3. Limites de controle

**Aproximação de Box (qui‑quadrado escalonada):**
$$g = \frac{v}{2m},\quad h = \frac{2m^2}{v},\quad Q_{\lim} = g \cdot \chi^2_{h,\alpha}$$
onde $m$ e $v$ são média e variância dos SPE do treinamento.

**Aproximação de Jackson‑Mudholkar (usada no código):**
$$Q_{\lim} = \theta_1 \left[ \frac{c_\alpha \sqrt{2\theta_2 h_0^2}}{\theta_1} + 1 + \frac{\theta_2 h_0 (h_0-1)}{\theta_1^2} \right]^{1/h_0}$$
com $\theta_i = \sum_{j=l+1}^{p} \lambda_j^i$ e $h_0 = 1 - \frac{2\theta_1\theta_3}{3\theta_2^2}$.
```

## 3.1. Algorítmo PLS

Quando se utiliza um modelo **PLS** (Mínimos Quadrados Parciais) para monitorar um processo com múltiplas entradas ($\mathbf{X}$) e múltiplas saídas ($\mathbf{Y}$), as estatísticas **$T^2$ de Hotelling** e **$Q$ (SPE – Squared Prediction Error)** permitem detectar desvios em relação ao comportamento normal. Ambas são calculadas a partir dos escores do modelo construído com dados de referência.

Para $N$ observações normais, o PLS decompõe as matrizes da seguinte forma:

$$\mathbf{X} = \mathbf{T} \mathbf{P}^\top + \mathbf{E}$$

$$\mathbf{Y} = \mathbf{T} \mathbf{Q}^\top + \mathbf{F}$$

* **$\mathbf{T}$** ($N \times A$): Matriz de escores (scores), ortogonais entre si.
* **$\mathbf{P}$** e **$\mathbf{Q}$**: Loadings de $\mathbf{X}$ e $\mathbf{Y}$, respectivamente.
* **$A$**: Número de componentes latentes selecionadas.

Para uma **nova amostra** $(\mathbf{x}_{\text{new}}, \mathbf{y}_{\text{new}})$, obtemos os escores projetados:

$$\mathbf{t}_{\text{new}} = \mathbf{x}_{\text{new}}^\top \mathbf{R}$$


onde $\mathbf{R} = \mathbf{W}(\mathbf{P}^\top \mathbf{W})^{-1}$ representa a matriz de pesos de projeção.


### 3.1.1. Estratégia de Monitoramento

Utilizando as estatísticas $T^2$ e $Q$ para reconstrução de x e y, para cada nova amostra, comparamos os índices com os limites calculados na fase de treino:

1. **$T^2 \uparrow$, $Q$ normal:** O processo mudou de regime (ex: aumento de produção), mas os sensores continuam correlacionados.
2. **$Q \uparrow$, $T^2$ normal:** Falha técnica ou anomalia estrutural (ex: sensor travado ou vazamento).
3. **Ambos $\uparrow$:** Anomalia grave e fora do espaço de operação conhecido.
   

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cross_decomposition import PLSRegression
from sklearn.preprocessing import StandardScaler
from scipy.stats import chi2
import warnings

warnings.filterwarnings("ignore")

# =====================================================================
# 1. FUNÇÕES AUXILIARES MATEMÁTICAS E GRÁFICAS
# =====================================================================

def calcular_metricas_completas(X_sc, Y_sc, pls_model, cov_T):
    """Calcula T2, Q_X (para as entradas) e Q_Y (para a saída)."""
    T = pls_model.transform(X_sc)
    T2 = np.zeros(X_sc.shape[0])
    inv_cov = np.linalg.inv(cov_T)
    for i in range(X_sc.shape[0]):
        T2[i] = T[i, :].T @ inv_cov @ T[i, :]
        
    X_pred = T @ pls_model.x_loadings_.T
    E_x = X_sc - X_pred
    Q_x = np.sum(E_x**2, axis=1)
    
    Y_pred = pls_model.predict(X_sc)
    E_y = Y_sc - Y_pred
    Q_y = np.sum(E_y**2, axis=1)
    
    return T2, Q_x, Q_y, E_x, E_y

def plotar_monitoramento(df_res, limites, confianca):
    """Plota T2, Q_X e Q_Y em 3 painéis usando gráficos do tipo stem."""
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 10))

    # T2 - Azul com marcador circular ('bo')
    ax1.stem(df_res.index, df_res['T2'], linefmt='b-', markerfmt='bo', basefmt='k-', label='T²')
    ax1.axhline(y=limites['T2'], color='red', linestyle='--', label=f'Limite {confianca*100}%')
    ax1.set_title('Controle T² (Desvios Globais em X)')
    ax1.legend()

    # Q_X - Laranja (C1) com marcador quadrado ('s')
    ax2.stem(df_res.index, df_res['Q_X'], linefmt='C1-', markerfmt='C1s', basefmt='k-', label='Q_X')
    ax2.axhline(y=limites['Q_X'], color='red', linestyle='--', label=f'Limite {confianca*100}%')
    ax2.set_title('Controle Q_X (Erros/Quebras de Correlação nas Entradas X)')
    ax2.legend()

    # Q_Y - Verde com marcador triangular ('g^')
    ax3.stem(df_res.index, df_res['Q_Y'], linefmt='g-', markerfmt='g^', basefmt='k-', label='Q_Y')
    ax3.axhline(y=limites['Q_Y'], color='red', linestyle='--', label=f'Limite {confianca*100}%')
    ax3.set_title('Controle Q_Y (Erro de Predição na Qualidade Y)')
    ax3.set_xlabel('Índice da Amostra')
    ax3.legend()

    plt.tight_layout()
    plt.show()
def plotar_contribuicao_qy(amostra_idx, E_y, colunas_y):
    """Mostra qual variável de SAÍDA (Y) gerou o maior erro de predição."""
    if isinstance(colunas_y, str): colunas_y = [colunas_y]
    contribuicoes = E_y[amostra_idx] ** 2
    
    plt.figure(figsize=(10, 5))
    plt.bar(colunas_y, contribuicoes, color='purple')
    plt.title(f'Contribuição das variáveis Y para o Erro Q_Y (Amostra {amostra_idx})')
    plt.ylabel('Contribuição (Erro Quadrático)')
    plt.xlabel('Variáveis de Saída (Y)')
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--')
    plt.tight_layout()
    plt.show()

def plotar_influencia_x_no_y(amostra_idx, X_sc, pls_model, colunas_x, y_idx=0):
    """Mostra o peso (influência) de cada variável de ENTRADA (X) na predição do valor de Y."""
    if isinstance(colunas_x, str): colunas_x = [colunas_x]
    
    coeficientes = pls_model.coef_[:, y_idx]
    influencia = X_sc[amostra_idx] * coeficientes
    
    plt.figure(figsize=(12, 5))
    cores = ['green' if val > 0 else 'red' for val in influencia]
    plt.bar(colunas_x, influencia, color=cores)
    plt.title(f'Influência das variáveis X na predição de Y (Amostra {amostra_idx})')
    plt.ylabel('Impacto na Predição (Valor X * Coef PLS)')
    plt.xlabel('Variáveis de Entrada (X)')
    plt.axhline(0, color='black', linewidth=1)
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--')
    plt.tight_layout()
    plt.show()

def printar_anomalias(resultados):
    total_amostras = len(resultados)
    qtd_t2 = resultados['Alarme_T2'].sum()
    qtd_qx = resultados['Alarme_Q_X'].sum()
    qtd_qy = resultados['Alarme_Q_Y'].sum()
    
    print(f"\n=== RESUMO DE ANOMALIAS (Teste com {total_amostras} amostras) ===")
    print(f"Alarmes T²  (Desvios globais em X)        : {qtd_t2}")
    print(f"Alarmes Q_X (Quebra de correlação em X)   : {qtd_qx}")
    print(f"Alarmes Q_Y (Erro na predição da Qualidade): {qtd_qy}")
    print("==============================================================\n")

# =====================================================================
# 2. WRAPPER 1: TREINAMENTO OFFLINE
# =====================================================================

def treinar_pls_offline(df_treino, colunas_x, colunas_y, n_components=1, confianca=0.95):
    """Treina o modelo e calcula os limites estatísticos, retornando um 'pacote' congelado."""
    if isinstance(colunas_x, str): colunas_x = [colunas_x]
    if isinstance(colunas_y, str): colunas_y = [colunas_y]
    
    X = df_treino[colunas_x]
    Y = df_treino[colunas_y]
    
    scaler_X = StandardScaler()
    scaler_Y = StandardScaler()
    X_train_sc = scaler_X.fit_transform(X)
    Y_train_sc = scaler_Y.fit_transform(Y)
    
    pls = PLSRegression(n_components=n_components)
    pls.fit(X_train_sc, Y_train_sc)
    
    cov_T = np.cov(pls.x_scores_.T)
    if n_components == 1:
        cov_T = np.array([[cov_T]])
        
    # Calcula métricas apenas para definir os limites
    _, Q_x_train, Q_y_train, _, _ = calcular_metricas_completas(X_train_sc, Y_train_sc, pls, cov_T)
    
    limites = {
        'T2': chi2.ppf(confianca, df=n_components),
        'Q_X': np.percentile(Q_x_train, confianca * 100),
        'Q_Y': np.percentile(Q_y_train, confianca * 100),
        'confianca': confianca
    }
    
    # Empacota os parâmetros para uso futuro
    modelo_offline = {
        'pls': pls,
        'scaler_X': scaler_X,
        'scaler_Y': scaler_Y,
        'cov_T': cov_T,
        'limites': limites,
        'colunas_x': colunas_x,
        'colunas_y': colunas_y
    }
    print("Modelo PLS treinado offline com sucesso!")
    return modelo_offline

# =====================================================================
# 3. WRAPPER 2: APLICAÇÃO ONLINE (Monitoramento) 
# =====================================================================
def aplicar_pls_online(df_teste, modelo_offline, gerar_graficos=True):
    """Usa o modelo congelado em novos dados. Gera plots e retorna as anomalias separadas."""
    # Desempacota o modelo
    pls = modelo_offline['pls']
    scaler_X = modelo_offline['scaler_X']
    scaler_Y = modelo_offline['scaler_Y']
    cov_T = modelo_offline['cov_T']
    limites = modelo_offline['limites']
    colunas_x = modelo_offline['colunas_x']
    colunas_y = modelo_offline['colunas_y']
    confianca = limites['confianca']
    
    # Escalonamento apenas usando TRANSFORM (congelado)
    X_test_sc = scaler_X.transform(df_teste[colunas_x])
    Y_test_sc = scaler_Y.transform(df_teste[colunas_y])
    
    # Calcula as métricas online
    T2_test, Q_x_test, Q_y_test, E_x_test, E_y_test = calcular_metricas_completas(X_test_sc, Y_test_sc, pls, cov_T)
    
    df_resultados = pd.DataFrame({
        'T2': T2_test,
        'Q_X': Q_x_test,
        'Q_Y': Q_y_test,
        'Alarme_T2': T2_test > limites['T2'],
        'Alarme_Q_X': Q_x_test > limites['Q_X'],
        'Alarme_Q_Y': Q_y_test > limites['Q_Y']
    }, index=df_teste.index)
    
    df_resultados['Posicao_Array'] = range(len(df_resultados))
    
    # Mapear o que é anomalia geral
    df_resultados['Anomalia'] = df_resultados['Alarme_T2'] | df_resultados['Alarme_Q_X'] | df_resultados['Alarme_Q_Y']
    
    # ISOLAR AS AMOSTRAS RUINS (T² , Q_x ou Q_y estourados)
    df_anomalias = df_resultados[df_resultados['Anomalia']].copy()
    
    # Plotagem (se ativado)
    if gerar_graficos:
        plotar_monitoramento(df_resultados, limites, confianca)
        
    printar_anomalias(df_resultados)
    
    # Retorna o df_anomalias como o 5º elemento!
    return df_resultados, E_x_test, E_y_test, X_test_sc, df_anomalias
````

## 3.1. PCA




### 3.1.1. Objetivo

O PCA (Análise de Componentes Principais) é um método de **redução de dimensionalidade não supervisionada** que transforma um conjunto de variáveis possivelmente correlacionadas em novas variáveis não correlacionadas (ortogonais), chamadas **componentes principais**. Essas componentes são ordenadas de forma que as primeiras retenham a maior parte da variância presente nos dados originais.

No monitoramento de processos, o PCA é utilizado para modelar o comportamento normal das variáveis de processo. A partir desse modelo, calculam‑se as estatísticas T² e Q, que permitem detectar desvios em relação à condição normal de operação — indicando a ocorrência de falhas —, mesmo sem conhecimento prévio dos modos de falha.

### 3.1.2. Formulação Matemática

Dada a matriz de dados centrados (e opcionalmente normalizados) **X** $(n \times p)$, o PCA decompõe a matriz de covariância:

$$ \mathbf{C} = \frac{1}{n-1} \mathbf{X}^\top \mathbf{X} $$

Resolvendo o problema de autovalores:

$$ \mathbf{C} \mathbf{p}_i = \lambda_i \mathbf{p}_i $$

Os autovetores $\mathbf{p}_i$ formam a matriz de **loadings** $\mathbf{P}$ ($p \times p$). Os autovalores $\lambda_i$ representam a variância explicada por cada componente. As componentes principais (escores) são obtidas projetando os dados nas direções dos autovetores:

$$ \mathbf{T} = \mathbf{X} \mathbf{P}_a $$

onde $\mathbf{P}_a$ ($p \times a$) contém as $a$ primeiras colunas de $\mathbf{P}$, correspondentes aos maiores autovalores. O número de componentes $a$ é escolhido por critérios como variância acumulada (ex.: 90%) ou validação cruzada.

A reconstrução da observação original e o resíduo são:

$$ \hat{\mathbf{X}} = \mathbf{T} \mathbf{P}_a^\top $$
$$ \mathbf{E} = \mathbf{X} - \hat{\mathbf{X}} $$

A implementação pode ser feita via SVD:

$$ \mathbf{X} = \mathbf{U} \mathbf{\Sigma} \mathbf{V}^\top $$
$$ \mathbf{P} = \mathbf{V}, \quad \mathbf{T} = \mathbf{U} \mathbf{\Sigma} $$

### 3.1.3. Limites de Controle

Assumindo normalidade dos escores e resíduos (aproximação válida para grande parte dos processos):

- **Limite T²** – Distribuição F:
  $$ T^2_{\lim} = \frac{a(n-1)}{n-a} F_{\alpha}(a,\; n-a) $$
  com $n$ amostras de treino e $a$ componentes retidas.

- **Limite Q** – Aproximação de Jackson‑Mudholkar:
  $$ Q_{\lim} = \theta_1 \left[ \frac{c_\alpha \sqrt{2\theta_2 h_0^2}}{\theta_1} + 1 + \frac{\theta_2 h_0 (h_0-1)}{\theta_1^2} \right]^{1/h_0} $$
  onde:
  - $\theta_k = \sum_{i=a+1}^p \lambda_i^k$ (autovalores descartados),
  - $h_0 = 1 - \frac{2\theta_1 \theta_3}{3\theta_2^2}$,
  - $c_\alpha$ = quantil da normal padrão.

### Aplicação no Processo Tennessee Eastman

Para o nosso caso, utilizamos como entrada do PCA as mesmas variáveis de processo disponíveis online:

- **X** : 33 variáveis (22 XMEAS contínuas + 11 XMV manipuladas)

O modelo PCA é treinado exclusivamente com os dados normais (`d00.dat`) e, em seguida, aplicado aos dados de teste (normal ou com falhas). As estatísticas T² e Q são monitoradas ao longo do tempo; violações dos limitefalhas que afetem o comportamento dinâmico do sistema, e não apenas os valores instantâneos.

## 3.2. DPCA - PCA Dinâmico

### 3.2.1. Objetivo

O DPCA (PCA Dinâmico) é uma extensão do PCA projetada para capturar **relações temporais** (autocorrelações) presentes em processos dinâmicos. Enquanto o PCA estático modela apenas a correlação instantânea entre variáveis, o DPCA incorpora observações passadas na matriz de entrada, permitindo que o modelo represente o comportamento dinâmico do sistema.

### 3.2.2. Formulação Matemática

Constrói‑se uma matriz de dados aumentada com **l** defasagens temporais:

$$ \mathbf{X}_{\text{lag}} = \left[ \mathbf{X}(k) \; \mathbf{X}(k-1) \; \dots \; \mathbf{X}(k-l) \right] $$

onde $\mathbf{X}(k-d)$ denota a matriz original deslocada $d$ instantes no tempo. A dimensão de $\mathbf{X}_{\text{lag}}$ é $(N-l) \times p(l+1)$, com $N$ amostras originais.

Em seguida, aplica‑se o PCA tradicional sobre $\mathbf{X}_{\text{lag}}$:

- Matriz de covariância: $\mathbf{C}_{\text{lag}} = \frac{1}{N-l-1} \mathbf{X}_{\text{lag}}^\top \mathbf{X}_{\text{lag}}$
- Decomposição em autovalores/autovetores: $\mathbf{C}_{\text{lag}} \mathbf{p}_i = \lambda_i \mathbf{p}_i$
- Escores: $\mathbf{T} = \mathbf{X}_{\text{lag}} \mathbf{P}_a$

O número de defasagens $l$ pode ser determinado por:
- Análise da autocorrelação dos escores do PCA estático,
- Método de variância reconstruída,
- Critério de informação (AIC, BIC) aplicado a modelos auto‑regressivos dos escores.

Durante a monitoração, cada nova amostra exige formar o vetor aumentado com as últimas $l+1$ observações, projetá‑lo no espaço de componentes e calcular os resíduos como no PCA estático.

### 3.2.3. Limites de Controle

Os limites são os mesmos do PCA estático, aplicados aos escores e resíduos obtidos a partir da matriz aumentada. As fórmulas de T² e Q permanecem idênticas; apenas o número de amostras efetivas de treino ($N-l$) e a dimensão do espaço residual ($p(l+1)-a$) mudam.

### Aplicação no Processo Tennessee Eastman

No processo TE, muitas variáveis exibem dinâmica significativa (inércia de sensores, atuadores). O DPCA é treinado com o mesmo conjunto de dados normais (`d00.dat`), usando as 33 variáveis de processo e, por exemplo, $l=1$ ou $2$ defasagens (ajustável conforme análise). A monitoração segue o mesmo princípio: T² e Q são calculados para cada instante e comparados com os limites, oferecendo maior sensibilidade a falhas que afetem o comportamento dinâmico do sistema, e não apenas os valores instantâneos.

````python
# ===========================================================
# Definição da classe para calculo do DPCA em lotes
# ==========================================================
class DPCAFaultDetector:
    def __init__(self, n_components=15, alpha=0.01, lags=1):
        """
        Dynamic PCA (DPCA) baseado no SVD para detecção de falhas.

        * n_components: Quantidade de valores singulares retidos (l).
          -> DICA: Como o DPCA multiplica o número de variáveis, 
             geralmente n_components precisa ser MAIOR que no PCA comum.
        * alpha: Tolerância teórica para alarmes falsos.
        * lags: Número de atrasos temporais (Ex: lags=1 pega t e t-1).
        """
        self.l = n_components
        self.alpha = alpha
        self.lags = lags
        self.scaler = StandardScaler()

        # Parâmetros aprendidos do modelo no treino (fit)
        self.P_pc = None
        self.P_res = None
        self.Lamb_pc_inv = None
        self.J_th_SPE = None 
        self.J_th_T2  = None 

    def _augment_data(self, X):
        """
        Método interno para criar a matriz aumentada com atrasos (lags).
        Se X tem 'm' variáveis, a saída terá m * (lags + 1) variáveis.
        """
        X_np = np.array(X)
        N, m = X_np.shape
        
        # Cria a nova matriz de dados aumentada
        X_aug = np.zeros((N, m * (self.lags + 1)))
        
        for lag in range(self.lags + 1):
            start_col = lag * m
            end_col = (lag + 1) * m
            
            if lag == 0:
                X_aug[:, start_col:end_col] = X_np
            else:
                # Desloca os dados para baixo em 'lag' posições (instantes passados)
                X_aug[lag:, start_col:end_col] = X_np[:-lag, :]
                # Padding: preenche os primeiros instantes sem histórico com o primeiro valor 
                # (Assume-state steady-state no instante inicial para não perder amostras)
                X_aug[:lag, start_col:end_col] = X_np[0, :]
                
        return X_aug

    def fit(self, X_train):
        """
        Treinar o modelo DPCA com os dados de funcionamento normal.
        """
        # 1. Aumenta os dados com os lags ANTES de normalizar
        X_train_aug = self._augment_data(X_train)
        
        # 2. Normalizar os dados de treino aumentados
        Z_train = self.scaler.fit_transform(X_train_aug)
        N = Z_train.shape[0]

        # 3. Matriz de covariância e SVD
        cov_matriz = np.cov(Z_train, rowvar=False)
        P, lamb, _ = np.linalg.svd(cov_matriz)

        # 4. Matriz dos autovalores e autovetores escolhidos
        self.P_pc  = P[:, :self.l]
        self.P_res = P[:, self.l:]
        
        Lamb_pc = np.diag(lamb[:self.l])
        self.Lamb_pc_inv = np.linalg.inv(Lamb_pc)

        # 5. Valores de theta para calculo do limite da estatística Q (SPE)
        theta_1 = np.sum(lamb[self.l:]**1)
        theta_2 = np.sum(lamb[self.l:]**2)
        theta_3 = np.sum(lamb[self.l:]**3)

        # Cálculo dos Limites
        h0 = 1 - (2 * theta_1 * theta_3) / (3 * theta_2**2)
        
        c_alpha = norm.ppf(1 - self.alpha)
        F_alpha = f.ppf(1 - self.alpha, dfn=self.l, dfd=(N - self.l))

        # Limite SPE (Q)
        self.J_th_SPE = theta_1 * ((c_alpha * np.sqrt(2 * theta_2 * h0**2)) / theta_1 + 1 + (theta_2 * h0 * (h0 - 1)) / theta_1**2)**(1 / h0)
        
        # Limite T²
        self.J_th_T2 = ((self.l * (N**2 - 1)) / (N * (N - self.l))) * F_alpha

        # DEBUG
        print(f"--- Treinamento DPCA Concluído ---")
        print(f"Lags utilizados: {self.lags} | Variáveis Aumentadas: {X_train_aug.shape[1]}")
        print(f"Componentes (l): {self.l} | Alpha: {self.alpha}")
        print(f"J_th_SPE (Limite Q) : {self.J_th_SPE:.4f}")
        print(f"J_th_T2  (Limite T²): {self.J_th_T2:.4f}\n")


    def test_batch(self, X_test):
        """
        Aplicar o modelo em um novo lote de dados.
        Retorno: estatísticas SPE, T2 e a decisão de falha (0 - Normal, 1 - Falha)
        """
        # 1. Aumenta os dados de teste com os lags
        X_test_aug = self._augment_data(X_test)
        
        # 2. Normaliza usando o scaler treinado no FIT
        Z_test = self.scaler.transform(X_test_aug)

        # 3. Aplicando o cálculo vetorizado
        E = Z_test @ self.P_res 
        SPE = np.sum(E**2, axis=1) 

        # T2 (Espaço principal)
        T = Z_test @ self.P_pc 
        T2 = np.sum((T @ self.Lamb_pc_inv) * T, axis=1) 

        # Previsão da falha lógica
        y_pred = ((SPE > self.J_th_SPE) | (T2 > self.J_th_T2)).astype(int)

        return SPE, T2, y_pred
        
    def plot_statistics(self, SPE, T2,id=10):
        """
        Plota os gráficos de SPE e T2 para um lote específico.
        """
        plt.figure(figsize=(16, 5))
        
        # Gráfico SPE
        plt.subplot(1, 2, 1)
        plt.plot(SPE[id], 'b', linewidth=2, label='SPE')
        plt.axhline(self.J_th_SPE, color='r', linestyle='--', linewidth=2, label='SPE Threshold')
        plt.title(f'SPE IPV[{id}] usando DPCA')
        plt.xlabel('Amostras')
        plt.ylabel('SPE')
        plt.legend()
        plt.grid()

        # Gráfico T2
        plt.subplot(1, 2, 2)
        plt.plot(T2[id], 'b', linewidth=2, label='T2')
        plt.axhline(self.J_th_T2, color='r', linestyle='--', linewidth=2, label='T2 Threshold')
        plt.title(f'T2 IPV[{id}] usando DPCA')
        plt.xlabel('Amostras')
        plt.ylabel('T2')
        plt.legend()
        plt.grid()
        
        plt.tight_layout()
        plt.show()
````

## 3.3. PLS (Partial Least Squares)


### 3.3.1. Objetivo

O PLS é um método de **redução de dimensionalidade supervisionada** que modela a relação entre duas matrizes de dados:

- **X** : variáveis de processo (entrada) – dimensão $(n \times p)$
- **Y** : variáveis de qualidade (saída) – dimensão $(n \times q)$

Seu objetivo é encontrar um conjunto de **variáveis latentes** (ou componentes) que **maximizem a covariância entre projeções de X e Y**, ou seja, extrair direções em X que melhor expliquem a variabilidade de Y.

No contexto de monitoramento de processos, usamos o PLS para construir um modelo que relaciona as condições operacionais (**X**) com a qualidade do produto (**Y**). Porém, como a qualidade geralmente só é medida em laboratório (de forma offline), a monitoração **online** é feita apenas no espaço de entrada **X**, utilizando as estatísticas T² e Q derivadas do modelo PLS.


### 3.3.2. Formulação Matemática

O PLS decompõe X e Y em:

- **X** = T Pᵀ + E  
- **Y** = U Qᵀ + F  

Onde:
- **T** ($n \times a$) : escores (variáveis latentes) de X
- **U** ($n \times a$) : escores de Y
- **P** ($p \times a$) : pesos (loadings) de X
- **Q** ($q \times a$) : pesos (loadings) de Y
- **E**, **F** : resíduos

O número de componentes $a$ é um hiperparâmetro, normalmente escolhido por validação cruzada ou por variância explicada.

O algoritmo mais comum para calcular PLS é o **NIPALS** (Nonlinear Iterative Partial Least Squares), que extrai uma componente por vez maximizando:

$$ \max_{\mathbf{w},\mathbf{c}} \; \text{cov}(\mathbf{X w},\; \mathbf{Y c}) $$

sujeito a $||\mathbf{w}|| = 1$, $||\mathbf{c}|| = 1$.

Para cada componente:

1. **t** = X **w**  
2. **u** = Y **c** (inicialmente u = uma coluna de Y)  
3. Repetir até convergência:  
   - **w** ∝ Xᵀ **u**  
   - **t** = X **w** , normalizar t  
   - **c** ∝ Yᵀ **t**  
   - **u** = Y **c**  
4. Deflacionar X e Y:  
   - **p** = Xᵀ **t** / (tᵀ t)  
   - X ← X – t pᵀ  
   - Y ← Y – t cᵀ (ou usando um vetor de loadings de Y)

Alternativamente, o PLS pode ser obtido via **SVD** da matriz de covariância cruzada **XᵀY** – as direções de máxima covariância são os vetores singulares.


### 3.3.3. Limites de Controle

Assumindo normalidade dos dados de treino:

- **Limite T²** – segue uma distribuição F:
  $$ T^2_{\lim} = \frac{a(n-1)}{n-a} F_{\alpha}(a,\; n-a) $$
  com $n$ = número de amostras de treino, $a$ = número de componentes PLS.

- **Limite Q** – pode ser aproximado por:
  $$ Q_{\lim} = \theta_1 \left[ \frac{c_\alpha \sqrt{2\theta_2 h_0^2}}{\theta_1} + 1 + \frac{\theta_2 h_0 (h_0-1)}{\theta_1^2} \right]^{1/h_0} $$
  onde:
  - $\theta_k = \sum_{i=a+1}^p \lambda_i^k$ ($\lambda_i$ são os autovalores da matriz de covariância dos resíduos de X, obtidos via PCA dos resíduos),
  - $h_0 = 1 - \frac{2\theta_1 \theta_3}{3\theta_2^2}$,
  - $c_\alpha$ é o quantil normal padrão $(1-\alpha)$.

Na prática, podemos também usar o percentil empírico do Q do conjunto de treino, especialmente se a suposição de normalidade não for válida.


### Aplicação no Processo Tennessee Eastman

No nosso caso, definimos:

- **X** : 33 variáveis (22 XMEAS contínuas + 11 XMV manipuladas)
- **Y** : XMEAS(35) – componente G no produto (qualidade)

Após treinar o PLS com os dados normais (d00), monitoramos as mesmas 33 variáveis de processo online usando T² e Q. Um aumento súbito indica uma possível falha no processo, mesmo que a qualidade ainda não tenha sido medida.

> **Nota**: O PLS também pode ser usado para predizer a qualidade offline (por exemplo, em amostras de laboratório), mas a detecção de falhas é baseada exclusivamente nos resíduos e escores de X.

````python
class PLSFaultDetector:
    def __init__(self, n_components=9, alpha=0.01):
        """
        Monitoramento de Falhas baseado em PLS (Partial Least Squares).
        
        :param n_components: Número de variáveis latentes retidas.
        :param alpha: Tolerância para alarmes falsos (ex: 0.01 para 99% de confiança).
        """
        self.l = n_components
        self.alpha = alpha
        
        # Scalers separados para X e Y
        self.scaler_X = StandardScaler()
        self.scaler_Y = StandardScaler()
        
        # O modelo PLS (com scale=False pois nós faremos a normalização manualmente)
        self.pls_model = PLSRegression(n_components=self.l, scale=False)
        
        # Parâmetros aprendidos
        self.var_T = None # Variância dos scores de treino para cálculo do T2
        self.J_th_T2  = None
        self.J_th_Qx  = None
        self.J_th_Qy  = None

    def _calculate_Q_limit(self, Q_train):
        """
        Calcula o limite para a estatística Q usando a Aproximação de Box (Distribuição Qui-Quadrado).
        """
        m = np.mean(Q_train)
        v = np.var(Q_train)
        g = v / (2 * m)
        h = (2 * m**2) / v
        limit = g * chi2.ppf(1 - self.alpha, df=h)
        return limit

    def fit(self, X_train, Y_train):
        """
        Treina o modelo PLS e calcula os limites das estatísticas de falha.
        """
        # 1. Normalização
        X_s = self.scaler_X.fit_transform(X_train)
        Y_s = self.scaler_Y.fit_transform(Y_train)
        N = X_s.shape[0]

        # 2. Treina o modelo PLS
        self.pls_model.fit(X_s, Y_s)
        
        # 3. Extração das matrizes do PLS
        T = self.pls_model.x_scores_
        P = self.pls_model.x_loadings_
        
        # 4. Cálculo da variância dos Scores (para o T2)
        # Como os scores T no NIPALS são ortogonais, a matriz de covariância é diagonal.
        self.var_T = np.var(T, axis=0) 
        
        # Limite T2 de Hotelling
        F_alpha = f.ppf(1 - self.alpha, dfn=self.l, dfd=(N - self.l))
        self.J_th_T2 = ((self.l * (N**2 - 1)) / (N * (N - self.l))) * F_alpha

        # 5. Reconstrução e cálculo dos Resíduos de Treino (para limites Q)
        # Resíduo de X (Qx)
        X_hat = T @ P.T
        E = X_s - X_hat
        Qx_train = np.sum(E**2, axis=1)
        
        # Resíduo de Y (Qy - Erro de Predição)
        Y_hat = self.pls_model.predict(X_s)
        F_res = Y_s - Y_hat
        Qy_train = np.sum(F_res**2, axis=1)

        # 6. Limites de Q usando Aproximação de Box
        self.J_th_Qx = self._calculate_Q_limit(Qx_train)
        self.J_th_Qy = self._calculate_Q_limit(Qy_train)

        # DEBUG
        print(f"--- Treinamento PLS Concluído ---")
        print(f"Componentes (l): {self.l} | Alpha: {self.alpha}")
        print(f"J_th_T2 (Limite T²): {self.J_th_T2:.4f}")
        print(f"J_th_Qx (Limite Q_x): {self.J_th_Qx:.4f}")
        print(f"J_th_Qy (Limite Q_y): {self.J_th_Qy:.4f}\n")

    def test_batch(self, X_test, Y_test):
        """
        Testa um novo lote de dados.
        Retorna as estatísticas T2, Qx, Qy e a decisão de falha (0 - Normal, 1 - Falha).
        """
        # 1. Normaliza usando os scalers do treino
        X_s = self.scaler_X.transform(X_test)
        Y_s = self.scaler_Y.transform(Y_test)

        # 2. Projeta X no espaço de scores usando a matriz de rotação do PLS (R)
        # R = W (P^T W)^-1. O Sklearn guarda isso em x_rotations_
        R = self.pls_model.x_rotations_
        T_test = X_s @ R

        # 3. Cálculo do T² (Vetorizado)
        # Divide os scores ao quadrado pela variância aprendida no treino e soma as colunas
        T2 = np.sum((T_test**2) / self.var_T, axis=1)

        # 4. Cálculo do Qx (Resíduos de Entrada)
        P = self.pls_model.x_loadings_
        X_hat = T_test @ P.T
        E = X_s - X_hat
        Qx = np.sum(E**2, axis=1)

        # 5. Cálculo do Qy (Resíduos de Saída / Erro de Predição)
        Y_hat = self.pls_model.predict(X_s)
        F_res = Y_s - Y_hat
        Qy = np.sum(F_res**2, axis=1)

        # 6. Decisão de Falha Lógica
        # Se QUALQUER UMA das estatísticas romper o limite, sinaliza falha.
        y_pred = ((T2 > self.J_th_T2) | (Qx > self.J_th_Qx) | (Qy > self.J_th_Qy)).astype(int)

        return T2, Qx, Qy, y_pred

    def plot_statistics(self, T2, Qx, Qy, id=10):
        """
        Plota os três gráficos de monitoramento: T2, Qx e Qy.
        """
        plt.figure(figsize=(20, 5))
        
        # Gráfico T2 (Scores)
        plt.subplot(1, 3, 1)
        plt.plot(T2[id], 'b', linewidth=1.5, label='T2')
        plt.axhline(self.J_th_T2, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'T² IVP[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('T²')
        plt.legend()
        plt.grid()

        # Gráfico Qx (Entrada)
        plt.subplot(1, 3, 2)
        plt.plot(Qx[id], 'g', linewidth=1.5, label='Qx (SPE Entrada)')
        plt.axhline(self.J_th_Qx, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'Qx IVP[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('Qx')
        plt.legend()
        plt.grid()
        
        # Gráfico Qy (Saída)
        plt.subplot(1, 3, 3)
        plt.plot(Qy[id], 'purple', linewidth=1.5, label='Qy (SPE Saída)')
        plt.axhline(self.J_th_Qy, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'Qy IVP[{id}] (Erro de Predição)')
        plt.xlabel('Amostras')
        plt.ylabel('Qy')
        plt.legend()
        plt.grid()
        
        plt.tight_layout()
        plt.show()
````
