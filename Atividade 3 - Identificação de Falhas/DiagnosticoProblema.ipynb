| Unidade | Função principal |
|---------|------------------|

# 2. Definição dos Teste de Hipótese que serão utilizados

## 2.1 Estatística $T^2$ (Hotelling)
### 2.2.1. O que mede

A **distância dos escores** da nova observação ao centro dos escores de treino, dentro do subespaço latente. Detecta variações **sistemáticas** (mudanças no ponto de operação) que ainda respeitam a estrutura de correlação do modelo.

### 2.2.2. Fórmula

$$T^2 = \mathbf{t}_{\text{new}}^\top \mathbf{S}^{-1} \mathbf{t}_{\text{new}} = \sum_{a=1}^{A} \frac{t_{\text{new},a}^2}{\lambda_a}$$

* $\mathbf{S} = \frac{\mathbf{T}_{\text{treino}}^\top \mathbf{T}_{\text{treino}}}{N-1}$ (matriz de covariância dos escores).
* $\lambda_a = \text{Var}(t_a)$ no conjunto de treinamento.

### 2.2.3. Limite de controle

$$T^2_{\lim} = \frac{A\,(N^2 - 1)}{N\,(N - A)}\; F_{A,\, N-A,\; \alpha}$$


onde $F$ é o quantil da distribuição de Fisher com nível de significância $\alpha$ (ex: 0.95 ou 0.99).

> **Interpretação:** $T^2$ alto $\rightarrow$ A amostra está longe da média, mas o modelo ainda "entende" a relação entre as variáveis.

## 2.3 Estatística SPE (Squared Prediction Error) - Estatística Q

### 2.3.1. O que mede

A variabilidade **não explicada** pelo modelo. Detecta quebras na estrutura de correlação do processo (ex.: falhas de sensores, mudanças dinâmicas).

### 2.3.2. Formulação
Para uma nova amostra $\mathbf{x}_{\text{new}}$:
- Projeção no espaço principal: $\mathbf{t}_{\text{new}} = \mathbf{x}_{\text{new}} \mathbf{P}$
- Resíduo: $\mathbf{e} = \mathbf{x}_{\text{new}} - \mathbf{t}_{\text{new}} \mathbf{P}^\top$
- **SPE** = $\mathbf{e}\mathbf{e}^\top = \|\mathbf{e}\|^2$

No código:
```python
SPE = z_test[i].T @ P_res @ P_res.T @ z_test[i]
```

### 2.3.3. Limites de controle

**Aproximação de Box (qui‑quadrado escalonada):**
$$g = \frac{v}{2m},\quad h = \frac{2m^2}{v},\quad Q_{\lim} = g \cdot \chi^2_{h,\alpha}$$
onde $m$ e $v$ são média e variância dos SPE do treinamento.

**Aproximação de Jackson‑Mudholkar (usada no código):**
$$Q_{\lim} = \theta_1 \left[ \frac{c_\alpha \sqrt{2\theta_2 h_0^2}}{\theta_1} + 1 + \frac{\theta_2 h_0 (h_0-1)}{\theta_1^2} \right]^{1/h_0}$$
com $\theta_i = \sum_{j=l+1}^{p} \lambda_j^i$ e $h_0 = 1 - \frac{2\theta_1\theta_3}{3\theta_2^2}$.

## 3.1. PCA




### 3.1.1. Objetivo

O PCA (Análise de Componentes Principais) é um método de **redução de dimensionalidade não supervisionada** que transforma um conjunto de variáveis possivelmente correlacionadas em novas variáveis não correlacionadas (ortogonais), chamadas **componentes principais**. Essas componentes são ordenadas de forma que as primeiras retenham a maior parte da variância presente nos dados originais.

No monitoramento de processos, o PCA é utilizado para modelar o comportamento normal das variáveis de processo. A partir desse modelo, calculam‑se as estatísticas T² e Q, que permitem detectar desvios em relação à condição normal de operação — indicando a ocorrência de falhas —, mesmo sem conhecimento prévio dos modos de falha.

### 3.1.2. Formulação Matemática

Dada a matriz de dados centrados (e opcionalmente normalizados) **X** $(n \times p)$, o PCA decompõe a matriz de covariância:

$$ \mathbf{C} = \frac{1}{n-1} \mathbf{X}^\top \mathbf{X} $$

Resolvendo o problema de autovalores:

$$ \mathbf{C} \mathbf{p}_i = \lambda_i \mathbf{p}_i $$

Os autovetores $\mathbf{p}_i$ formam a matriz de **loadings** $\mathbf{P}$ ($p \times p$). Os autovalores $\lambda_i$ representam a variância explicada por cada componente. As componentes principais (escores) são obtidas projetando os dados nas direções dos autovetores:

$$ \mathbf{T} = \mathbf{X} \mathbf{P}_a $$

onde $\mathbf{P}_a$ ($p \times a$) contém as $a$ primeiras colunas de $\mathbf{P}$, correspondentes aos maiores autovalores. O número de componentes $a$ é escolhido por critérios como variância acumulada (ex.: 90%) ou validação cruzada.

A reconstrução da observação original e o resíduo são:

$$ \hat{\mathbf{X}} = \mathbf{T} \mathbf{P}_a^\top $$
$$ \mathbf{E} = \mathbf{X} - \hat{\mathbf{X}} $$

A implementação pode ser feita via SVD:

$$ \mathbf{X} = \mathbf{U} \mathbf{\Sigma} \mathbf{V}^\top $$
$$ \mathbf{P} = \mathbf{V}, \quad \mathbf{T} = \mathbf{U} \mathbf{\Sigma} $$

### 3.1.3. Limites de Controle

Assumindo normalidade dos escores e resíduos (aproximação válida para grande parte dos processos):

- **Limite T²** – Distribuição F:
  $$ T^2_{\lim} = \frac{a(n-1)}{n-a} F_{\alpha}(a,\; n-a) $$
  com $n$ amostras de treino e $a$ componentes retidas.

- **Limite Q** – Aproximação de Jackson‑Mudholkar:
  $$ Q_{\lim} = \theta_1 \left[ \frac{c_\alpha \sqrt{2\theta_2 h_0^2}}{\theta_1} + 1 + \frac{\theta_2 h_0 (h_0-1)}{\theta_1^2} \right]^{1/h_0} $$
  onde:
  - $\theta_k = \sum_{i=a+1}^p \lambda_i^k$ (autovalores descartados),
  - $h_0 = 1 - \frac{2\theta_1 \theta_3}{3\theta_2^2}$,
  - $c_\alpha$ = quantil da normal padrão.

O modelo PCA é treinado exclusivamente com os dados normais (`d00.dat`) e, em seguida, aplicado aos dados de teste (normal ou com falhas). As estatísticas T² e Q são monitoradas ao longo do tempo; violações dos limitefalhas que afetem o comportamento dinâmico do sistema, e não apenas os valores instantâneos.

````python
# ===========================================================
# Definição da classe para calculo do PCA/SVD em lotes
# ==========================================================
class PCAFaultDetector:
    def __init__(self, n_components=9, alpha = 0.01):
        """
            A ideia é utilizar um modelo PCA baseado no SVD para detecção de falhas

            *n_components: Quantidade de valores singulares
            *alpha: Tolerância para alarmes falsos
        """
        self.l = n_components
        self.alpha = alpha
        self.scaler = StandardScaler() #Para normalizar os dados

        # Parâmetros aprendidos do modelo no treino (fit)
        self.P_pc = None
        self.Lamb_pc_inv = None
        self.J_th_SPE = None #Limiar para Q
        self.J_th_T2  = None #Limiar para T²

    def fit(self,X_train):
        """
            Treinar o modelo com os dados escolhidos para o treino
        """
        # Normalizar os dados de treino
        Z_train = self.scaler.fit_transform(X_train)
        N = Z_train.shape[0]  #Número de valores

        # Matriz de covariância e SVD
        cov_matriz = np.cov(Z_train, rowvar=False)
        P, lamb, _ = svd(cov_matriz)

        
        # Matriz dos autovalores e autovetores escolhidos (os l componentes)
        self.P_pc  = P[:, :self.l]
        self.P_res = P[:, self.l:]
        
        Lamb_pc = np.diag(lamb[:self.l])
        self.Lamb_pc_inv = np.linalg.inv(Lamb_pc)

        # Valores de theta para calculo do limite da estatística Q (SPE)
        theta_1 = np.sum(lamb[self.l:]**1)
        theta_2 = np.sum(lamb[self.l:]**2)
        theta_3 = np.sum(lamb[self.l:]**3)

        # Cálculo dos Limites
        h0 = 1 - (2 * theta_1 * theta_3) / (3 * theta_2**2)
        
        c_alpha = norm.ppf(1 - self.alpha)
        F_alpha = f.ppf(1 - self.alpha, dfn=self.l, dfd=(N - self.l))

        ## Cálculo dos limites com base nos valores de treino
        # Limite SPE (Q)
        self.J_th_SPE = theta_1 * ((c_alpha * np.sqrt(2 * theta_2 * h0**2)) / theta_1 + 1 + (theta_2 * h0 * (h0 - 1)) / theta_1**2)**(1 / h0)
        
        # Limite T²
        self.J_th_T2 = ((self.l * (N**2 - 1)) / (N * (N - self.l))) * F_alpha

        # DEBUG
        print(f"--- Treinamento Concluído ---")
        print(f"Componentes (l): {self.l} | Alpha: {self.alpha}")
        print(f"J_th_SPE (Limite Q) : {self.J_th_SPE:.4f}")
        print(f"J_th_T2  (Limite T²): {self.J_th_T2:.4f}\n")


    def test_batch(self,X_test):
        """
        Aplicar o modelo em um novo lote de dados.

        Retorno: estatísticas SPE, T2 e a decisão de falha 
        Decisão de falha:
            0 - Normal
            1 - Falha
        """
        # Normalizo os dados
        Z_test = self.scaler.transform(X_test)

        # Aplicando o cálculo de forma vetorizada (no batch completo)
        E = Z_test @ self.P_res #Projeto todo o espaço no espaço residual (dos resíduos).
        SPE = np.sum(E**2, axis=1) #Soma o quadrado dos resíduos.

        # T2 (Espaço principal)
        T = Z_test @ self.P_pc #Projeta no espaço principal (scores)
        T2 = np.sum((T@self.Lamb_pc_inv)*T, axis =1 ) #T2 de Hotelling vetorizado

        # Previsão da falha lógica (SPE e T2 devem superar o limite)
        y_pred = ((SPE > self.J_th_SPE) | (T2>self.J_th_T2)).astype(int)

        return SPE, T2, y_pred
        
    def plot_statistics(self, SPE, T2, id=10):
        """
        Plota os gráficos de SPE e T2 para um lote específico.
        """
        plt.figure(figsize=(16, 5))
        
        # Gráfico SPE
        plt.subplot(1, 2, 1)
        plt.plot(SPE[id], 'b', linewidth=2, label='SPE')
        plt.axhline(self.J_th_SPE, color='r', linestyle='--', linewidth=2, label='SPE Threshold')
        plt.title(f'SPE IPV[{id}] usando PCA')
        plt.xlabel('Amostras')
        plt.ylabel('SPE')
        plt.legend()
        plt.grid()

        # Gráfico T2
        plt.subplot(1, 2, 2)
        plt.plot(T2[id], 'b', linewidth=2, label='T2')
        plt.axhline(self.J_th_T2, color='r', linestyle='--', linewidth=2, label='T2 Threshold')
        plt.title(f'T2 IPV[{id}] usando PCA')
        plt.xlabel('Amostras')
        plt.ylabel('T2')
        plt.legend()
        plt.grid()
        
        plt.tight_layout()
        plt.show()
````

````python
# =============================================================
# Selecionando os dados de Treino
#==============================================================
# Definição das colunas
xmeas = [f"XMEAS({i})" for i in range(1, 23)]
xmv   = [f"XMV({i})" for i in range(1, 12)]
colunas = xmeas+xmv 

# Coletando os dados para treinar o modelo
dados_treino = df_CN_treino[colunas]
dados_teste  = df_testes_completo

# Implementar e treinar o modelo
pca_monitor = PCAFaultDetector(n_components=9, alpha=0.01)
pca_monitor.fit(dados_treino)

print(len(dados_teste))
````

## 3.2. DPCA - PCA Dinâmico

### 3.2.1. Objetivo

O DPCA (PCA Dinâmico) é uma extensão do PCA projetada para capturar **relações temporais** (autocorrelações) presentes em processos dinâmicos. Enquanto o PCA estático modela apenas a correlação instantânea entre variáveis, o DPCA incorpora observações passadas na matriz de entrada, permitindo que o modelo represente o comportamento dinâmico do sistema.

### 3.2.2. Formulação Matemática

Constrói‑se uma matriz de dados aumentada com **l** defasagens temporais:

$$ \mathbf{X}_{\text{lag}} = \left[ \mathbf{X}(k) \; \mathbf{X}(k-1) \; \dots \; \mathbf{X}(k-l) \right] $$

onde $\mathbf{X}(k-d)$ denota a matriz original deslocada $d$ instantes no tempo. A dimensão de $\mathbf{X}_{\text{lag}}$ é $(N-l) \times p(l+1)$, com $N$ amostras originais.

Em seguida, aplica‑se o PCA tradicional sobre $\mathbf{X}_{\text{lag}}$:

- Matriz de covariância: $\mathbf{C}_{\text{lag}} = \frac{1}{N-l-1} \mathbf{X}_{\text{lag}}^\top \mathbf{X}_{\text{lag}}$
- Decomposição em autovalores/autovetores: $\mathbf{C}_{\text{lag}} \mathbf{p}_i = \lambda_i \mathbf{p}_i$
- Escores: $\mathbf{T} = \mathbf{X}_{\text{lag}} \mathbf{P}_a$

O número de defasagens $l$ pode ser determinado por:
- Análise da autocorrelação dos escores do PCA estático,
- Método de variância reconstruída,
- Critério de informação (AIC, BIC) aplicado a modelos auto‑regressivos dos escores.

Durante a monitoração, cada nova amostra exige formar o vetor aumentado com as últimas $l+1$ observações, projetá‑lo no espaço de componentes e calcular os resíduos como no PCA estático.

### 3.2.3. Limites de Controle

Os limites são os mesmos do PCA estático, aplicados aos escores e resíduos obtidos a partir da matriz aumentada. As fórmulas de T² e Q permanecem idênticas; apenas o número de amostras efetivas de treino ($N-l$) e a dimensão do espaço residual ($p(l+1)-a$) mudam.

### Aplicação no Processo Tennessee Eastman

No processo TE, muitas variáveis exibem dinâmica significativa (inércia de sensores, atuadores). O DPCA é treinado com o mesmo conjunto de dados normais (`d00.dat`), usando as 33 variáveis de processo e, por exemplo, $l=1$ ou $2$ defasagens (ajustável conforme análise). A monitoração segue o mesmo princípio: T² e Q são calculados para cada instante e comparados com os limites, oferecendo maior sensibilidade a falhas que afetem o comportamento dinâmico do sistema, e não apenas os valores instantâneos.

````python
# ===========================================================
# Definição da classe para calculo do DPCA em lotes
# ==========================================================
class DPCAFaultDetector:
    def __init__(self, n_components=15, alpha=0.01, lags=1):
        """
        Dynamic PCA (DPCA) baseado no SVD para detecção de falhas.

        * n_components: Quantidade de valores singulares retidos (l).
          -> DICA: Como o DPCA multiplica o número de variáveis, 
             geralmente n_components precisa ser MAIOR que no PCA comum.
        * alpha: Tolerância teórica para alarmes falsos.
        * lags: Número de atrasos temporais (Ex: lags=1 pega t e t-1).
        """
        self.l = n_components
        self.alpha = alpha
        self.lags = lags
        self.scaler = StandardScaler()

        # Parâmetros aprendidos do modelo no treino (fit)
        self.P_pc = None
        self.P_res = None
        self.Lamb_pc_inv = None
        self.J_th_SPE = None 
        self.J_th_T2  = None 

    def _augment_data(self, X):
        """
        Método interno para criar a matriz aumentada com atrasos (lags).
        Se X tem 'm' variáveis, a saída terá m * (lags + 1) variáveis.
        """
        X_np = np.array(X)
        N, m = X_np.shape
        
        # Cria a nova matriz de dados aumentada
        X_aug = np.zeros((N, m * (self.lags + 1)))
        
        for lag in range(self.lags + 1):
            start_col = lag * m
            end_col = (lag + 1) * m
            
            if lag == 0:
                X_aug[:, start_col:end_col] = X_np
            else:
                # Desloca os dados para baixo em 'lag' posições (instantes passados)
                X_aug[lag:, start_col:end_col] = X_np[:-lag, :]
                # Padding: preenche os primeiros instantes sem histórico com o primeiro valor 
                # (Assume-state steady-state no instante inicial para não perder amostras)
                X_aug[:lag, start_col:end_col] = X_np[0, :]
                
        return X_aug

    def fit(self, X_train):
        """
        Treinar o modelo DPCA com os dados de funcionamento normal.
        """
        # 1. Aumenta os dados com os lags ANTES de normalizar
        X_train_aug = self._augment_data(X_train)
        
        # 2. Normalizar os dados de treino aumentados
        Z_train = self.scaler.fit_transform(X_train_aug)
        N = Z_train.shape[0]

        # 3. Matriz de covariância e SVD
        cov_matriz = np.cov(Z_train, rowvar=False)
        P, lamb, _ = np.linalg.svd(cov_matriz)

        # 4. Matriz dos autovalores e autovetores escolhidos
        self.P_pc  = P[:, :self.l]
        self.P_res = P[:, self.l:]
        
        Lamb_pc = np.diag(lamb[:self.l])
        self.Lamb_pc_inv = np.linalg.inv(Lamb_pc)

        # 5. Valores de theta para calculo do limite da estatística Q (SPE)
        theta_1 = np.sum(lamb[self.l:]**1)
        theta_2 = np.sum(lamb[self.l:]**2)
        theta_3 = np.sum(lamb[self.l:]**3)

        # Cálculo dos Limites
        h0 = 1 - (2 * theta_1 * theta_3) / (3 * theta_2**2)
        
        c_alpha = norm.ppf(1 - self.alpha)
        F_alpha = f.ppf(1 - self.alpha, dfn=self.l, dfd=(N - self.l))

        # Limite SPE (Q)
        self.J_th_SPE = theta_1 * ((c_alpha * np.sqrt(2 * theta_2 * h0**2)) / theta_1 + 1 + (theta_2 * h0 * (h0 - 1)) / theta_1**2)**(1 / h0)
        
        # Limite T²
        self.J_th_T2 = ((self.l * (N**2 - 1)) / (N * (N - self.l))) * F_alpha

        # DEBUG
        print(f"--- Treinamento DPCA Concluído ---")
        print(f"Lags utilizados: {self.lags} | Variáveis Aumentadas: {X_train_aug.shape[1]}")
        print(f"Componentes (l): {self.l} | Alpha: {self.alpha}")
        print(f"J_th_SPE (Limite Q) : {self.J_th_SPE:.4f}")
        print(f"J_th_T2  (Limite T²): {self.J_th_T2:.4f}\n")


    def test_batch(self, X_test):
        """
        Aplicar o modelo em um novo lote de dados.
        Retorno: estatísticas SPE, T2 e a decisão de falha (0 - Normal, 1 - Falha)
        """
        # 1. Aumenta os dados de teste com os lags
        X_test_aug = self._augment_data(X_test)
        
        # 2. Normaliza usando o scaler treinado no FIT
        Z_test = self.scaler.transform(X_test_aug)

        # 3. Aplicando o cálculo vetorizado
        E = Z_test @ self.P_res 
        SPE = np.sum(E**2, axis=1) 

        # T2 (Espaço principal)
        T = Z_test @ self.P_pc 
        T2 = np.sum((T @ self.Lamb_pc_inv) * T, axis=1) 

        # Previsão da falha lógica
        y_pred = ((SPE > self.J_th_SPE) | (T2 > self.J_th_T2)).astype(int)

        return SPE, T2, y_pred
        
    def plot_statistics(self, SPE, T2,id=10):
        """
        Plota os gráficos de SPE e T2 para um lote específico.
        """
        plt.figure(figsize=(16, 5))
        
        # Gráfico SPE
        plt.subplot(1, 2, 1)
        plt.plot(SPE[id], 'b', linewidth=2, label='SPE')
        plt.axhline(self.J_th_SPE, color='r', linestyle='--', linewidth=2, label='SPE Threshold')
        plt.title(f'SPE IPV[{id}] usando DPCA')
        plt.xlabel('Amostras')
        plt.ylabel('SPE')
        plt.legend()
        plt.grid()

        # Gráfico T2
        plt.subplot(1, 2, 2)
        plt.plot(T2[id], 'b', linewidth=2, label='T2')
        plt.axhline(self.J_th_T2, color='r', linestyle='--', linewidth=2, label='T2 Threshold')
        plt.title(f'T2 IPV[{id}] usando DPCA')
        plt.xlabel('Amostras')
        plt.ylabel('T2')
        plt.legend()
        plt.grid()
        
        plt.tight_layout()
        plt.show()
````

````python
# =============================================================
# Selecionando os dados de Treino
#==============================================================
# Definição das colunas
xmeas = [f"XMEAS({i})" for i in range(1, 23)]
xmv   = [f"XMV({i})" for i in range(1, 12)]
colunas = xmeas+xmv 

# Coletando os dados para treinar o modelo
dados_treino = df_CN_treino[colunas]
dados_teste  = df_testes_completo

# Implementar e treinar o modelo
dpca_monitor = DPCAFaultDetector(n_components=9, alpha=0.01)
dpca_monitor.fit(dados_treino)
````

## 3.3. PLS (Partial Least Squares)


### 3.3.1. Objetivo

O PLS é um método de **redução de dimensionalidade supervisionada** que modela a relação entre duas matrizes de dados:

- **X** : variáveis de processo (entrada) – dimensão $(n \times p)$
- **Y** : variáveis de qualidade (saída) – dimensão $(n \times q)$

Seu objetivo é encontrar um conjunto de **variáveis latentes** (ou componentes) que **maximizem a covariância entre projeções de X e Y**, ou seja, extrair direções em X que melhor expliquem a variabilidade de Y.

No contexto de monitoramento de processos, usamos o PLS para construir um modelo que relaciona as condições operacionais (**X**) com a qualidade do produto (**Y**). Porém, como a qualidade geralmente só é medida em laboratório (de forma offline), a monitoração **online** é feita apenas no espaço de entrada **X**, utilizando as estatísticas T² e Q derivadas do modelo PLS.


### 3.3.2. Formulação Matemática

O PLS decompõe X e Y em:

- **X** = T Pᵀ + E  
- **Y** = U Qᵀ + F  

Onde:
- **T** ($n \times a$) : escores (variáveis latentes) de X
- **U** ($n \times a$) : escores de Y
- **P** ($p \times a$) : pesos (loadings) de X
- **Q** ($q \times a$) : pesos (loadings) de Y
- **E**, **F** : resíduos

O número de componentes $a$ é um hiperparâmetro, normalmente escolhido por validação cruzada ou por variância explicada.

O algoritmo mais comum para calcular PLS é o **NIPALS** (Nonlinear Iterative Partial Least Squares), que extrai uma componente por vez maximizando:

$$ \max_{\mathbf{w},\mathbf{c}} \; \text{cov}(\mathbf{X w},\; \mathbf{Y c}) $$

sujeito a $||\mathbf{w}|| = 1$, $||\mathbf{c}|| = 1$.

Para cada componente:

1. **t** = X **w**  
2. **u** = Y **c** (inicialmente u = uma coluna de Y)  
3. Repetir até convergência:  
   - **w** ∝ Xᵀ **u**  
   - **t** = X **w** , normalizar t  
   - **c** ∝ Yᵀ **t**  
   - **u** = Y **c**  
4. Deflacionar X e Y:  
   - **p** = Xᵀ **t** / (tᵀ t)  
   - X ← X – t pᵀ  
   - Y ← Y – t cᵀ (ou usando um vetor de loadings de Y)

Alternativamente, o PLS pode ser obtido via **SVD** da matriz de covariância cruzada **XᵀY** – as direções de máxima covariância são os vetores singulares.


### 3.3.3. Limites de Controle

Assumindo normalidade dos dados de treino:

- **Limite T²** – segue uma distribuição F:
  $$ T^2_{\lim} = \frac{a(n-1)}{n-a} F_{\alpha}(a,\; n-a) $$
  com $n$ = número de amostras de treino, $a$ = número de componentes PLS.

- **Limite Q** – pode ser aproximado por:
  $$ Q_{\lim} = \theta_1 \left[ \frac{c_\alpha \sqrt{2\theta_2 h_0^2}}{\theta_1} + 1 + \frac{\theta_2 h_0 (h_0-1)}{\theta_1^2} \right]^{1/h_0} $$
  onde:
  - $\theta_k = \sum_{i=a+1}^p \lambda_i^k$ ($\lambda_i$ são os autovalores da matriz de covariância dos resíduos de X, obtidos via PCA dos resíduos),
  - $h_0 = 1 - \frac{2\theta_1 \theta_3}{3\theta_2^2}$,
  - $c_\alpha$ é o quantil normal padrão $(1-\alpha)$.

Na prática, podemos também usar o percentil empírico do Q do conjunto de treino, especialmente se a suposição de normalidade não for válida.


### Aplicação no Processo Tennessee Eastman

No nosso caso, definimos:

- **X** : 33 variáveis (22 XMEAS contínuas + 11 XMV manipuladas)
- **Y** : XMEAS(35) – componente G no produto (qualidade)

Após treinar o PLS com os dados normais (d00), monitoramos as mesmas 33 variáveis de processo online usando T² e Q. Um aumento súbito indica uma possível falha no processo, mesmo que a qualidade ainda não tenha sido medida.

> **Nota**: O PLS também pode ser usado para predizer a qualidade offline (por exemplo, em amostras de laboratório), mas a detecção de falhas é baseada exclusivamente nos resíduos e escores de X.

````python
class PLSFaultDetector:
    def __init__(self, n_components=9, alpha=0.01):
        """
        Monitoramento de Falhas baseado em PLS (Partial Least Squares).
        
        :param n_components: Número de variáveis latentes retidas.
        :param alpha: Tolerância para alarmes falsos (ex: 0.01 para 99% de confiança).
        """
        self.l = n_components
        self.alpha = alpha
        
        # Scalers separados para X e Y
        self.scaler_X = StandardScaler()
        self.scaler_Y = StandardScaler()
        
        # O modelo PLS (com scale=False pois nós faremos a normalização manualmente)
        self.pls_model = PLSRegression(n_components=self.l, scale=False)
        
        # Parâmetros aprendidos
        self.var_T = None # Variância dos scores de treino para cálculo do T2
        self.J_th_T2  = None
        self.J_th_Qx  = None
        self.J_th_Qy  = None

    def _calculate_Q_limit(self, Q_train):
        """
        Calcula o limite para a estatística Q usando a Aproximação de Box (Distribuição Qui-Quadrado).
        """
        m = np.mean(Q_train)
        v = np.var(Q_train)
        g = v / (2 * m)
        h = (2 * m**2) / v
        limit = g * chi2.ppf(1 - self.alpha, df=h)
        return limit

    def fit(self, X_train, Y_train):
        """
        Treina o modelo PLS e calcula os limites das estatísticas de falha.
        """
        # 1. Normalização
        X_s = self.scaler_X.fit_transform(X_train)
        Y_s = self.scaler_Y.fit_transform(Y_train)
        N = X_s.shape[0]

        # 2. Treina o modelo PLS
        self.pls_model.fit(X_s, Y_s)
        
        # 3. Extração das matrizes do PLS
        T = self.pls_model.x_scores_
        P = self.pls_model.x_loadings_
        
        # 4. Cálculo da variância dos Scores (para o T2)
        # Como os scores T no NIPALS são ortogonais, a matriz de covariância é diagonal.
        self.var_T = np.var(T, axis=0) 
        
        # Limite T2 de Hotelling
        F_alpha = f.ppf(1 - self.alpha, dfn=self.l, dfd=(N - self.l))
        self.J_th_T2 = ((self.l * (N**2 - 1)) / (N * (N - self.l))) * F_alpha

        # 5. Reconstrução e cálculo dos Resíduos de Treino (para limites Q)
        # Resíduo de X (Qx)
        X_hat = T @ P.T
        E = X_s - X_hat
        Qx_train = np.sum(E**2, axis=1)
        
        # Resíduo de Y (Qy - Erro de Predição)
        Y_hat = self.pls_model.predict(X_s)
        F_res = Y_s - Y_hat
        Qy_train = np.sum(F_res**2, axis=1)

        # 6. Limites de Q usando Aproximação de Box
        self.J_th_Qx = self._calculate_Q_limit(Qx_train)
        self.J_th_Qy = self._calculate_Q_limit(Qy_train)

        # DEBUG
        print(f"--- Treinamento PLS Concluído ---")
        print(f"Componentes (l): {self.l} | Alpha: {self.alpha}")
        print(f"J_th_T2 (Limite T²): {self.J_th_T2:.4f}")
        print(f"J_th_Qx (Limite Q_x): {self.J_th_Qx:.4f}")
        print(f"J_th_Qy (Limite Q_y): {self.J_th_Qy:.4f}\n")

    def test_batch(self, X_test, Y_test):
        """
        Testa um novo lote de dados.
        Retorna as estatísticas T2, Qx, Qy e a decisão de falha (0 - Normal, 1 - Falha).
        """
        # 1. Normaliza usando os scalers do treino
        X_s = self.scaler_X.transform(X_test)
        Y_s = self.scaler_Y.transform(Y_test)

        # 2. Projeta X no espaço de scores usando a matriz de rotação do PLS (R)
        # R = W (P^T W)^-1. O Sklearn guarda isso em x_rotations_
        R = self.pls_model.x_rotations_
        T_test = X_s @ R

        # 3. Cálculo do T² (Vetorizado)
        # Divide os scores ao quadrado pela variância aprendida no treino e soma as colunas
        T2 = np.sum((T_test**2) / self.var_T, axis=1)

        # 4. Cálculo do Qx (Resíduos de Entrada)
        P = self.pls_model.x_loadings_
        X_hat = T_test @ P.T
        E = X_s - X_hat
        Qx = np.sum(E**2, axis=1)

        # 5. Cálculo do Qy (Resíduos de Saída / Erro de Predição)
        Y_hat = self.pls_model.predict(X_s)
        F_res = Y_s - Y_hat
        Qy = np.sum(F_res**2, axis=1)

        # 6. Decisão de Falha Lógica
        # Se QUALQUER UMA das estatísticas romper o limite, sinaliza falha.
        y_pred = ((T2 > self.J_th_T2) | (Qx > self.J_th_Qx) | (Qy > self.J_th_Qy)).astype(int)

        return T2, Qx, Qy, y_pred

    def plot_statistics(self, T2, Qx, Qy, id=10):
        """
        Plota os três gráficos de monitoramento: T2, Qx e Qy.
        """
        plt.figure(figsize=(20, 5))
        
        # Gráfico T2 (Scores)
        plt.subplot(1, 3, 1)
        plt.plot(T2[id], 'b', linewidth=1.5, label='T2')
        plt.axhline(self.J_th_T2, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'T² IVP[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('T²')
        plt.legend()
        plt.grid()

        # Gráfico Qx (Entrada)
        plt.subplot(1, 3, 2)
        plt.plot(Qx[id], 'g', linewidth=1.5, label='Qx (SPE Entrada)')
        plt.axhline(self.J_th_Qx, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'Qx IVP[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('Qx')
        plt.legend()
        plt.grid()
        
        # Gráfico Qy (Saída)
        plt.subplot(1, 3, 3)
        plt.plot(Qy[id], 'purple', linewidth=1.5, label='Qy (SPE Saída)')
        plt.axhline(self.J_th_Qy, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'Qy IVP[{id}] (Erro de Predição)')
        plt.xlabel('Amostras')
        plt.ylabel('Qy')
        plt.legend()
        plt.grid()
        
        plt.tight_layout()
        plt.show()
````

## 3.5. TPLS

### 3.4.1. Objetivo

O **TPLS** é uma extensão do PLS convencional projetada especificamente para o monitoramento de processos. Enquanto o PLS padrão foca apenas na maximização da covariância entre $X$ e $Y$, o TPLS decompõe o espaço das variáveis de entrada ($X$) em subespaços ainda mais refinados:

1. **Subespaço Preditivo ($\hat{X}_p$):** Variações em $X$ que estão diretamente correlacionadas com as variações em $Y$ (qualidade).
2. **Subespaço Ortogonal ($\hat{X}_o$):** Variações em $X$ que são significativas (grande variância), mas que **não possuem impacto** nas variáveis de saída $Y$.
3. **Subespaço Residual ($E$):** Ruído e variações não capturadas pelos componentes anteriores.

O grande objetivo do TPLS é evitar **alarmes falsos**. Em muitos processos, uma variável de entrada pode mudar drasticamente (ex: uma oscilação comum de carga), mas se essa variação não afeta a qualidade do produto, ela não deveria ser tratada com a mesma severidade que uma falha que compromete o produto final. O TPLS permite separar o que é "falha de processo" do que é "falha de qualidade".

### 3.4.2. Formulação Matemática

O TPLS utiliza uma decomposição sequencial. Primeiro, extraímos a parte preditiva via PLS e, em seguida, aplicamos PCA nos resíduos de $X$ para extrair a parte ortogonal.

A decomposição de $X$ no TPLS é dada por:

$$\mathbf{X} = \underbrace{\mathbf{T}_p \mathbf{P}_p^T}_{\text{Preditivo}} + \underbrace{\mathbf{T}_o \mathbf{P}_o^T}_{\text{Ortogonal}} + \mathbf{E}$$

Onde:

* $\mathbf{T}_p$ e $\mathbf{P}_p$: Scores e Loadings obtidos pelo PLS (variáveis que explicam $Y$).
* $\mathbf{T}_o$ e $\mathbf{P}_o$: Scores e Loadings obtidos via PCA nos resíduos de $X$ (variáveis que não explicam $Y$, mas têm alta variância).
* $\mathbf{E}$: Matriz de resíduos finais.

Para a monitoração, definimos estatísticas para cada subespaço:

1. **$T^2_p$**: Monitora variações no espaço principal que afetam a qualidade.
2. **$T^2_o$**: Monitora variações no espaço principal que **não** afetam a qualidade.
3. **$Q$ (ou SPE)**: Monitora o erro de reconstrução (resíduos).

### 3.4.3. Limites de Controle

Assim como no PLS, utilizamos distribuições estatísticas para os limiares:

* **Limite $T^2$ ($T^2_p$ e $T^2_o$):** Calculado via distribuição F de Hotelling ou aproximado por Qui-Quadrado.
* **Limite $Q$:** Calculado via aproximação de Box (Qui-Quadrado), garantindo robustez aos resíduos.

$$J_{th} = g \cdot \chi^2_{\alpha}(h)$$


Onde $g$ e $h$ são parâmetros de ajuste baseados na média e variância das estatísticas calculadas no conjunto de treino.

### O que muda na prática com o TPLS?

Ao rodar esse código no Tennessee Eastman:

1. **T² Preditivo:** Mostrará falhas que afetam diretamente o `XMEAS(35)` (Qualidade G).
2. **T² Ortogonal:** Mostrará variações que são "grandes" nos sensores, mas que o sistema de controle talvez consiga compensar antes de afetar a qualidade final.
3. **Q:** Mostrará eventos repentinos ou ruídos que fogem totalmente do padrão histórico aprendido.

### Implementação: `TPLSFaultDetector`

Esta classe implementa a lógica do TPLS, separando a análise preditiva (PLS) da análise de variância ortogonal (PCA sobre os resíduos).


````python
class TPLSFaultDetector:
    def __init__(self, n_pls=5, n_pca=4, alpha=0.01):
        """
        Monitoramento de Falhas baseado em Total PLS (TPLS).
        
        :param n_pls: Componentes preditivos (relacionados a Y).
        :param n_pca: Componentes ortogonais (variações em X não relacionadas a Y).
        :param alpha: Nível de significância (ex: 0.01 para 99% de confiança).
        """
        self.n_pls = n_pls
        self.n_pca = n_pca
        self.alpha = alpha
        
        self.scaler_X = StandardScaler()
        self.scaler_Y = StandardScaler()
        
        self.pls = PLSRegression(n_components=n_pls, scale=False)
        self.pca_ortho = PCA(n_components=n_pca)
        
        # Limites
        self.J_th_T2p = None
        self.J_th_T2o = None
        self.J_th_Q   = None
        
        # Variâncias para normalização dos scores
        self.var_Tp = None
        self.var_To = None

    def _calculate_limit(self, data):
        """ Aproximação de Box para limites de controle """
        m = np.mean(data)
        v = np.var(data)
        g = v / (2 * m)
        h = (2 * m**2) / v
        return g * chi2.ppf(1 - self.alpha, df=h)

    def fit(self, X_train, Y_train):
        # 1. Normalização
        X_s = self.scaler_X.fit_transform(X_train)
        Y_s = self.scaler_Y.fit_transform(Y_train)
        N = X_s.shape[0]

        # 2. Passo PLS (Espaço Preditivo)
        self.pls.fit(X_s, Y_s)
        Tp = self.pls.x_scores_
        Pp = self.pls.x_loadings_
        self.var_Tp = np.var(Tp, axis=0)

        # 3. Cálculo do Resíduo Preditivo para o Passo PCA
        # X_hat_p = Tp @ Pp.T
        X_res_p = X_s - (Tp @ Pp.T)

        # 4. Passo PCA (Espaço Ortogonal - variações em X que PLS ignorou)
        self.pca_ortho.fit(X_res_p)
        To = self.pca_ortho.transform(X_res_p)
        self.var_To = np.var(To, axis=0)

        # 5. Resíduo Final (Q)
        Po = self.pca_ortho.components_.T
        X_res_final = X_res_p - (To @ Po.T)
        Q_train = np.sum(X_res_final**2, axis=1)

        # 6. Cálculo dos Limites
        # T2 Preditivo (usando F-dist ou Box)
        T2p_train = np.sum((Tp**2) / self.var_Tp, axis=1)
        self.J_th_T2p = self._calculate_limit(T2p_train)
        
        # T2 Ortogonal
        T2o_train = np.sum((To**2) / self.var_To, axis=1)
        self.J_th_T2o = self._calculate_limit(T2o_train)
        
        # Q Final
        self.J_th_Q = self._calculate_limit(Q_train)

        print(f"--- Treinamento TPLS Concluído ---")
        print(f"Limites: T2p={self.J_th_T2p:.2f}, T2o={self.J_th_T2o:.2f}, Q={self.J_th_Q:.2f}\n")

    def test_batch(self, X_test, Y_test):
        X_s = self.scaler_X.transform(X_test)
        
        # Projeção Preditiva (PLS)
        R = self.pls.x_rotations_
        Tp = X_s @ R
        T2p = np.sum((Tp**2) / self.var_Tp, axis=1)
        
        # Projeção Ortogonal (PCA nos resíduos do PLS)
        Pp = self.pls.x_loadings_
        X_res_p = X_s - (Tp @ Pp.T)
        To = self.pca_ortho.transform(X_res_p)
        T2o = np.sum((To**2) / self.var_To, axis=1)
        
        # Resíduo Final
        Po = self.pca_ortho.components_.T
        X_res_final = X_res_p - (To @ Po.T)
        Q = np.sum(X_res_final**2, axis=1)
        
        # Decisão: Falha se qualquer estatística exceder o limite
        y_pred = ((T2p > self.J_th_T2p) | (T2o > self.J_th_T2o) | (Q > self.J_th_Q)).astype(int)
        
        return T2p, T2o, Q, y_pred

    def plot_statistics(self, T2p, T2o, Q, id=0):
        plt.figure(figsize=(20, 5))
        
        stats = [T2p[id], T2o[id], Q[id]]
        limits = [self.J_th_T2p, self.J_th_T2o, self.J_th_Q]
        titles = ['T² Preditivo (Qualidade)', 'T² Ortogonal (Processo)', 'Q (Resíduos)']
        colors = ['b', 'g', 'm']

        for i in range(3):
            plt.subplot(1, 3, i+1)
            plt.plot(stats[i], color=colors[i], label='Estatística')
            plt.axhline(limits[i], color='r', linestyle='--', label='Limiar')
            plt.title(f'{titles[i]} - IDV[{id}]')
            plt.xlabel('Amostras')
            plt.legend()
            plt.grid(True)
        
        plt.tight_layout()
        plt.show()
````

## 3.5. MPLS

### 3.5.1. Objetivo

O **MPLS** é uma extensão do PLS projetada para monitorar processos de larga escala compostos por múltiplas unidades operacionais. Em vez de tratar todas as variáveis como um único bloco massivo, o MPLS divide a matriz $\mathbf{X}$ em sub-blocos significativos:

$$\mathbf{X} = [\mathbf{X}_1, \mathbf{X}_2, \dots, \mathbf{X}_b]$$

**Seus principais objetivos são:**

1. **Isolamento de Falhas:** Identificar não apenas *que* uma falha ocorreu, mas em qual unidade/bloco do processo ela se originou.
2. **Gerenciamento de Complexidade:** Facilitar a interpretação de modelos com centenas de variáveis.
3. **Monitoramento Hierárquico:** Fornecer estatísticas globais (do processo todo) e estatísticas locais (de cada bloco individual).

No Tennessee Eastman, isso permite separar as 22 medições contínuas e 11 variáveis manipuladas em blocos como "Alimentação", "Reator" e "Separação/Stippler".

### 3.5.2. Formulação Matemática

O MPLS utiliza o conceito de **super-escores** ($\mathbf{t}_T$) para combinar a informação de todos os blocos. A decomposição básica para cada bloco $i$ é:

$$\mathbf{X}_i = \mathbf{T}_i \mathbf{P}_i^T + \mathbf{E}_i$$

Onde a relação entre os blocos é capturada através de pesos de bloco ($\mathbf{w}_i$) que compõem o peso global $\mathbf{w}_T$. O algoritmo (geralmente uma variante do NIPALS) segue estes passos simplificados:

1. Extrair escores locais $\mathbf{t}_i$ de cada bloco $\mathbf{X}_i$.
2. Concatenar os escores locais para formar uma matriz de blocos $\mathbf{T}_B = [\mathbf{t}_1, \mathbf{t}_2, \dots, \mathbf{t}_b]$.
3. Calcular o **super-escore** $\mathbf{t}_T$ que maximiza a covariância com $\mathbf{Y}$.
4. Deflacionar as matrizes de bloco e repetir para o próximo componente.

### 3.5.3. Limites de Controle

No MPLS, o monitoramento é realizado em dois níveis:

1. **Nível Global:** Estatísticas $T^2_{global}$ e $Q_{global}$ calculadas a partir do modelo que integra todos os blocos.
2. **Nível de Bloco:** Estatísticas $T^2_{block, i}$ e $Q_{block, i}$ que indicam a saúde de cada unidade específica.

Os limites de controle seguem as mesmas premissas de distribuição do PLS convencional (Distribuição F e Aproximação de Box), mas são aplicados individualmente para cada estatística de bloco, permitindo uma "triagem" imediata da origem da falha.


### Implementação: `MPLSFaultDetector`

Esta implementação foca na capacidade de monitorar o processo global e fornecer uma visão detalhada por blocos.

````python
class MPLSFaultDetector:
    def __init__(self, block_indices, n_components=9, alpha=0.01):
        """
        Monitoramento de Falhas baseado em MPLS (Multi-Block PLS).
        
        :param block_indices: Lista de listas com índices de variáveis para cada bloco.
        :param n_components: Número de variáveis latentes retidas.
        :param alpha: Tolerância para alarmes falsos.
        """
        self.block_indices = block_indices
        self.n_blocks = len(block_indices)
        self.l = n_components
        self.alpha = alpha
        
        self.scaler_X = StandardScaler()
        self.scaler_Y = StandardScaler()
        self.pls = PLSRegression(n_components=n_components, scale=False)
        
        # Limites globais e por bloco
        self.J_th_T2_global = None
        self.J_th_Q_global = None
        self.J_th_Q_blocks = []  # Lista com limites Q para cada bloco
        
        # Variância dos scores para T2
        self.var_T_global = None

    def _box_limit(self, data):
        """ Limite para o SPE/Q usando a aproximação de Box (Qui-Quadrado) """
        m, v = np.mean(data), np.var(data)
        g, h = v / (2 * m), (2 * m**2) / v
        return g * chi2.ppf(1 - self.alpha, df=h)

    def _t2_limit(self, N):
        """ Limite exato para T2 baseado na distribuição F de Fisher """
        A = self.l
        F_val = f.ppf(1 - self.alpha, A, N - A)
        return (A * (N**2 - 1) / (N * (N - A))) * F_val

    def fit(self, X_train, Y_train):
        """
        Treina o modelo MPLS com dados normais.
        """
        N = X_train.shape[0]
        
        # 1. Normalização
        X_s = self.scaler_X.fit_transform(X_train)
        Y_s = self.scaler_Y.fit_transform(Y_train)
        
        # 2. Treinamento PLS Global (modelo global)
        self.pls.fit(X_s, Y_s)
        
        # 3. Extração das matrizes
        T_global = self.pls.x_scores_
        P_glob = self.pls.x_loadings_
        self.var_T_global = np.var(T_global, axis=0)
        
        # 4. Cálculo do Erro Global
        X_hat = T_global @ P_glob.T
        E_glob = X_s - X_hat
        Q_glob = np.sum(E_glob**2, axis=1)
        
        # 5. Limites Globais
        T2_glob = np.sum((T_global**2) / self.var_T_global, axis=1)
        self.J_th_T2_global = self._t2_limit(N)
        self.J_th_Q_global = self._box_limit(Q_glob)
        
        # 6. Limites por Bloco (Q de cada bloco)
        self.J_th_Q_blocks = []
        for i, indices in enumerate(self.block_indices):
            E_block = E_glob[:, indices]
            Q_block = np.sum(E_block**2, axis=1)
            limit_block = self._box_limit(Q_block)
            self.J_th_Q_blocks.append(limit_block)

        # DEBUG
        print(f"--- Treinamento MPLS Concluído ---")
        print(f"Blocos: {self.n_blocks} | Componentes: {self.l} | Alpha: {self.alpha}")
        print(f"J_th_T2 (Global): {self.J_th_T2_global:.4f}")
        print(f"J_th_Q (Global):  {self.J_th_Q_global:.4f}")
        for i, limit in enumerate(self.J_th_Q_blocks):
            print(f"J_th_Q (Bloco {i}): {limit:.4f}")
        print()

    def test_batch(self, X_test, Y_test=None):
        """
        Testa um novo lote de dados.
        Retorna T2 global, Q global, Q por bloco e predição.
        """
        X_s = self.scaler_X.transform(X_test)
        
        # Projeção no espaço de scores (usando x_rotations_)
        R = self.pls.x_rotations_
        P_glob = self.pls.x_loadings_
        
        # Estatísticas Globais
        T_glob = X_s @ R
        T2_glob = np.sum((T_glob**2) / self.var_T_global, axis=1)
        
        # Erro Global
        X_hat = T_glob @ P_glob.T
        E_glob = X_s - X_hat
        Q_glob = np.sum(E_glob**2, axis=1)
        
        # Estatísticas por Bloco (Q de cada bloco)
        Q_blocks = []
        for indices in self.block_indices:
            E_block = E_glob[:, indices]
            Q_b = np.sum(E_block**2, axis=1)
            Q_blocks.append(Q_b)
        
        # Decisão: Falha se qualquer estatística superar o limite
        y_pred = ((T2_glob > self.J_th_T2_global) | (Q_glob > self.J_th_Q_global)).astype(int)
        
        return T2_glob, Q_glob, Q_blocks, y_pred

    def plot_statistics(self, T2, Q, Q_blocks, id=0):
        """
        Plota as estatísticas de monitoramento: T2 global, Q global e Q por bloco.
        Configurado para uma grade estável de 2x3 (2 linhas e 3 colunas).
        """
        fig, axes = plt.subplots(2, 3, figsize=(18, 9))
        axes = axes.flatten()
        
        # 1. T2 Global
        axes[0].plot(T2, 'b', linewidth=1.5, label='T² Global')
        axes[0].axhline(self.J_th_T2_global, color='r', linestyle='--', linewidth=2, label='Limiar')
        axes[0].set_title(f'T² Global - IDV[{id}]')
        axes[0].set_ylabel('T²')
        axes[0].legend()
        axes[0].grid(True)
        
        # 2. Q Global
        axes[1].plot(Q, 'g', linewidth=1.5, label='Q Global')
        axes[1].axhline(self.J_th_Q_global, color='r', linestyle='--', linewidth=2, label='Limiar')
        axes[1].set_title(f'Q Global - IDV[{id}]')
        axes[1].set_ylabel('Q')
        axes[1].legend()
        axes[1].grid(True)
        
        # 3. Q por Bloco (Máximo 4 blocos para preencher os eixos 2, 3, 4 e 5)
        block_names = ['Reator', 'Separador', 'Stripper', 'Compressor']
        for i in range(min(self.n_blocks, 4)):
            idx = 2 + i
            axes[idx].plot(Q_blocks[i], linewidth=1.5, label=f'Q Bloco {i}')
            axes[idx].axhline(self.J_th_Q_blocks[i], color='r', linestyle='--', linewidth=2, label='Limiar')
            
            title = block_names[i] if i < len(block_names) else f'Bloco {i}'
            axes[idx].set_title(f'Q {title} - IDV[{id}]')
            axes[idx].set_ylabel('Q')
            axes[idx].legend()
            axes[idx].grid(True)
            
        # Configura o xlabel apenas para os gráficos que realmente foram plotados
        # E esconde os subplots que sobrarem (caso n_blocks < 4)
        for idx in range(6):
            if idx < 2 + min(self.n_blocks, 4):
                axes[idx].set_xlabel('Amostras')
            else:
                axes[idx].set_axis_off()  # Esconde o quadrado cinza/vazio se não houver bloco para ele
        
        plt.tight_layout()
        plt.show()
````

## 3.6. ICA (Independent Component Analysis)

### 3.6.1. Objetivo

O **ICA** (Análise de Componentes Independentes) é um método de redução de dimensionalidade que busca encontrar componentes **estatisticamente independentes** a partir de dados multivariados. Diferentemente do PCA, que encontra componentes com máxima variância, o ICA procura componentes que sejam **o máximo possível independentes** entre si, minimizando a informação mútua.

No monitoramento de processos, o ICA é útil para separar fontes de variação independentes (ex: diferentes modos de operação, tipos de falhas) que podem estar misturadas nos dados brutos.

### 3.6.2. Formulação Matemática

O modelo ICA assume que as observações $\mathbf{X}$ são uma mistura linear de componentes independentes:

$$\mathbf{X} = \mathbf{A} \mathbf{S}$$

Onde:
- **X** ($n \times p$) : dados observados (centrados)
- **A** ($p \times l$) : matriz de mistura (desconhecida)
- **S** ($l \times n$) : componentes independentes (desconhecidas)

O objetivo do ICA é estimar a matriz de separação $\mathbf{W} = \mathbf{A}^{-1}$ tal que:

$$\mathbf{S} = \mathbf{W} \mathbf{X}$$

Os componentes em $\mathbf{S}$ são maximamente independentes, normalmente medido pela **negentropy** ou **informação mútua**. O algoritmo mais comum é o **FastICA**, que utiliza uma função não-linear $g$ (ex: $\tanh$) para medir a não-gaussianidade.

### 3.6.3. Limites de Controle

Assim como no PCA, depois de obter os componentes independentes, calculamos as estatísticas **T²** (distância de Hotelling nos escores ICA) e **Q** (erro de reconstrução). Os limites seguem as mesmas premissas:

- **Limite T²:** Baseado na distribuição F com número de componentes e amostras de treino.
- **Limite Q:** Aproximação de Box (Qui-Quadrado escalonada).

### Aplicação no Processo Tennessee Eastman

O ICA é treinado com os mesmos 33 variáveis de entrada (22 XMEAS + 11 XMV), visando encontrar componentes independentes que capturem diferentes padrões de falhas.

````python
from sklearn.decomposition import FastICA

class ICAFaultDetector:
    def __init__(self, n_components=9, alpha=0.01, random_state=42):
        """
        Monitoramento de Falhas baseado em ICA (Independent Component Analysis).
        
        :param n_components: Número de componentes independentes.
        :param alpha: Tolerância para alarmes falsos.
        :param random_state: Seed para reprodutibilidade.
        """
        self.l = n_components
        self.alpha = alpha
        self.random_state = random_state
        
        self.scaler = StandardScaler()
        self.ica = FastICA(n_components=n_components, random_state=random_state, max_iter=500)
        
        # Parâmetros aprendidos
        self.A = None  # Matriz de mixing
        self.var_S = None  # Variância dos scores
        self.J_th_T2 = None
        self.J_th_Q = None

    def fit(self, X_train):
        """
        Treina o modelo ICA com dados de operação normal.
        """
        N = X_train.shape[0]
        
        # 1. Normalização
        Z_train = self.scaler.fit_transform(X_train)
        
        # 2. Treinamento ICA (FastICA)
        S_train = self.ica.fit_transform(Z_train)
        
        # 3. Extração da matriz de mistura
        self.A = self.ica.mixing_
        
        # 4. Cálculo da variância dos scores ICA
        self.var_S = np.var(S_train, axis=0)
        
        # 5. Cálculo dos Resíduos de Treino (Q)
        # Reconstrução: X_hat = S @ A.T
        X_hat = S_train @ self.A.T
        E = Z_train - X_hat
        Q_train = np.sum(E**2, axis=1)
        
        # 6. Cálculo dos Limites
        # T2 (Hotelling nos escores ICA)
        T2_train = np.sum((S_train**2) / self.var_S, axis=1)
        
        # Limite T2
        F_alpha = f.ppf(1 - self.alpha, dfn=self.l, dfd=(N - self.l))
        self.J_th_T2 = ((self.l * (N**2 - 1)) / (N * (N - self.l))) * F_alpha
        
        # Limite Q (Box)
        m, v = np.mean(Q_train), np.var(Q_train)
        g, h = v / (2 * m), (2 * m**2) / v
        self.J_th_Q = g * chi2.ppf(1 - self.alpha, df=h)
        
        # DEBUG
        print(f"--- Treinamento ICA Concluído ---")
        print(f"Componentes (l): {self.l} | Alpha: {self.alpha}")
        print(f"J_th_T2 (Limite T²): {self.J_th_T2:.4f}")
        print(f"J_th_Q  (Limite Q):  {self.J_th_Q:.4f}\n")

    def test_batch(self, X_test):
        """
        Aplica o modelo ICA em um novo lote de dados.
        """
        # 1. Normalização
        Z_test = self.scaler.transform(X_test)
        
        # 2. Projeção nos componentes ICA
        # S = X @ ica.components_.T (ou usar o fit_transform internamente)
        S_test = self.ica.transform(Z_test)
        
        # 3. Cálculo do T2
        T2 = np.sum((S_test**2) / self.var_S, axis=1)
        
        # 4. Cálculo do Q
        X_hat = S_test @ self.A.T
        E = Z_test - X_hat
        Q = np.sum(E**2, axis=1)
        
        # 5. Decisão de Falha
        y_pred = ((T2 > self.J_th_T2) | (Q > self.J_th_Q)).astype(int)
        
        return T2, Q, y_pred

    def plot_statistics(self, T2, Q, id=10):
        """
        Plota as estatísticas T2 e Q.
        """
        plt.figure(figsize=(16, 5))
        
        # Gráfico T2
        plt.subplot(1, 2, 1)
        plt.plot(T2[id], 'b', linewidth=2, label='T²')
        plt.axhline(self.J_th_T2, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'T² ICA - IDV[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('T²')
        plt.legend()
        plt.grid()
        
        # Gráfico Q
        plt.subplot(1, 2, 2)
        plt.plot(Q[id], 'g', linewidth=2, label='Q (SPE)')
        plt.axhline(self.J_th_Q, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'Q ICA - IDV[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('Q')
        plt.legend()
        plt.grid()
        
        plt.tight_layout()
        plt.show()
````

````python
# =============================================================
# Selecionando os dados de Treino para ICA
# =============================================================
# Definição das colunas (22 XMEAS + 11 XMV = 33 variáveis)
xmeas = [f"XMEAS({i})" for i in range(1, 23)]
xmv   = [f"XMV({i})" for i in range(1, 12)]
colunas = xmeas + xmv

# Coletando os dados para treinar o modelo
dados_treino = df_CN_treino[colunas]
dados_teste  = df_testes_completo

# Implementar e treinar o modelo ICA
ica_monitor = ICAFaultDetector(n_components=9, alpha=0.01)
ica_monitor.fit(dados_treino)

print(f"Dados preparados: treino={dados_treino.shape}, teste={len(dados_teste)}")
````

## 3.7. MICA (Multi-Block Independent Component Analysis)

### 3.7.1. Objetivo

O **MICA** (Análise de Componentes Independentes Multi-Bloco) é uma extensão do ICA que divide o espaço das variáveis em múltiplos blocos significativos (correspondentes a unidades ou seções do processo), realizando decomposição ICA para cada bloco individualmente e depois combinando os resultados em um modelo global.

O principal benefício é o **isolamento de falhas por bloco**: enquanto o ICA global encontra componentes independentes de todo o processo, o MICA adiciona uma camada de granularidade que permite identificar qual unidade específica está gerando comportamento anormal.

### 3.7.2. Formulação Matemática

O MICA estende o ICA padrão dividindo a matriz de dados em blocos:

$$\mathbf{X} = [\mathbf{X}_1, \mathbf{X}_2, \dots, \mathbf{X}_b]$$

Para cada bloco $i$, aplicamos ICA independentemente:

$$\mathbf{X}_i = \mathbf{A}_i \mathbf{S}_i$$

Os escores de todos os blocos são concatenados para formar uma **matriz de blocos de componentes**:

$$\mathbf{S}_B = [\mathbf{S}_1, \mathbf{S}_2, \dots, \mathbf{S}_b]$$

Opcionalmente, uma segunda rodada de ICA (ou PCA) pode ser aplicada a $\mathbf{S}_B$ para encontrar super-componentes que sintetizem a estrutura global.

### 3.7.3. Limites de Controle

O MICA fornece dois níveis de monitoramento:

1. **Nível Global:** Estatísticas T² e Q calculadas a partir dos componentes agregados.
2. **Nível de Bloco:** Cada bloco $i$ possui suas próprias estatísticas T²$_i$ e Q$_i$.

Os limites seguem as mesmas formulações: distribuição F para T² e aproximação de Box para Q.

### Aplicação no Processo Tennessee Eastman

O MICA divide o processo em 4 blocos (Reator, Separador, Stripper, Compressor), aplicando FastICA independentemente em cada bloco e combinando os resultados em um monitoramento global com isolamento por unidade.

````python
class MICAFaultDetector:
    def __init__(self, block_indices, n_components_block=3, alpha=0.01, random_state=42):
        """
        Monitoramento de Falhas baseado em MICA (Multi-Block ICA).
        
        :param block_indices: Lista de listas com índices de variáveis para cada bloco.
        :param n_components_block: Componentes ICA por bloco.
        :param alpha: Nível de significância.
        :param random_state: Seed para reprodutibilidade.
        """
        self.block_indices = block_indices
        self.n_blocks = len(block_indices)
        self.n_comp_block = n_components_block
        self.alpha = alpha
        self.random_state = random_state
        
        self.scalers = [StandardScaler() for _ in range(self.n_blocks)]
        self.icas = [FastICA(n_components=n_components_block, random_state=random_state, max_iter=500,whiten='unit-variance') 
                     for _ in range(self.n_blocks)]
        
        # Armazenar variâncias e limites
        self.var_S_blocks = []
        self.J_th_T2_blocks = []
        self.J_th_Q_blocks = []
        
        # Limites globais
        self.J_th_T2_global = None
        self.J_th_Q_global = None

    def _box_limit(self, data):
        m, v = np.mean(data), np.var(data)
        g, h = v / (2 * m), (2 * m**2) / v
        return g * chi2.ppf(1 - self.alpha, df=h)

    def _t2_limit(self, N, n_comp):
        F_val = f.ppf(1 - self.alpha, n_comp, N - n_comp)
        return (n_comp * (N**2 - 1) / (N * (N - n_comp))) * F_val

    def fit(self, X_train):
        """
        Treina MICA em cada bloco independentemente.
        """
        N = X_train.shape[0]
        
        # Processar cada bloco
        for b, indices in enumerate(self.block_indices):
            X_block = X_train[:, indices]
            
            # Normalização
            Z_block = self.scalers[b].fit_transform(X_block)
            
            # ICA
            S_block = self.icas[b].fit_transform(Z_block)
            var_S = np.var(S_block, axis=0)
            self.var_S_blocks.append(var_S)
            
            # Cálculo de resíduos
            A_block = self.icas[b].mixing_
            X_hat = S_block @ A_block.T
            E_block = Z_block - X_hat
            Q_block = np.sum(E_block**2, axis=1)
            
            # T2 do bloco
            T2_block = np.sum((S_block**2) / var_S, axis=1)
            
            # Limites
            th_T2 = self._t2_limit(N, self.n_comp_block)
            th_Q = self._box_limit(Q_block)
            
            self.J_th_T2_blocks.append(th_T2)
            self.J_th_Q_blocks.append(th_Q)
        
        # Limites globais (média dos blocos)
        self.J_th_T2_global = np.mean(self.J_th_T2_blocks)
        self.J_th_Q_global = np.mean(self.J_th_Q_blocks)
        
        # DEBUG
        print(f"--- Treinamento MICA Concluído ---")
        print(f"Blocos: {self.n_blocks} | Componentes/Bloco: {self.n_comp_block}")
        print(f"Limites Globais: T2={self.J_th_T2_global:.4f}, Q={self.J_th_Q_global:.4f}")
        for i in range(self.n_blocks):
            print(f"Bloco {i}: T2={self.J_th_T2_blocks[i]:.4f}, Q={self.J_th_Q_blocks[i]:.4f}")
        print()

    def test_batch(self, X_test):
        """
        Testa MICA e retorna estatísticas globais e por bloco.
        """
        T2_global_list = []
        Q_global_list = []
        Q_blocks = []
        y_pred_blocks = []
        
        n_samples = X_test.shape[0]
        
        for b, indices in enumerate(self.block_indices):
            X_block = X_test[:, indices]
            Z_block = self.scalers[b].transform(X_block)
            
            # Projeção ICA
            S_block = self.icas[b].transform(Z_block)
            var_S = self.var_S_blocks[b]
            
            # T2 e Q do bloco
            T2_block = np.sum((S_block**2) / var_S, axis=1)
            
            A_block = self.icas[b].mixing_
            X_hat = S_block @ A_block.T
            E_block = Z_block - X_hat
            Q_block = np.sum(E_block**2, axis=1)
            
            T2_global_list.append(T2_block)
            Q_global_list.append(Q_block)
            Q_blocks.append(Q_block)
            
            # Predição por bloco
            y_pred_b = ((T2_block > self.J_th_T2_blocks[b]) | (Q_block > self.J_th_Q_blocks[b])).astype(int)
            y_pred_blocks.append(y_pred_b)
        
        # Agregação global
        T2_global = np.mean(T2_global_list, axis=0)
        Q_global = np.mean(Q_global_list, axis=0)
        y_pred = ((T2_global > self.J_th_T2_global) | (Q_global > self.J_th_Q_global)).astype(int)
        
        return T2_global, Q_global, Q_blocks, y_pred

    def plot_statistics(self, T2_global, Q_global, Q_blocks, id=0):
        """
        Plota estatísticas globais e por bloco em uma grade.
        """
        block_names = ['Reator', 'Separador', 'Stripper', 'Compressor']
        
        # Como temos 6 gráficos (2 globais + 4 blocos), 
        # criamos uma grade 2x3 (2 linhas e 3 colunas).
        fig, axes = plt.subplots(2, 3, figsize=(18, 10))
        
        axes = axes.flatten()
        
        # T2 Global
        axes[0].plot(T2_global[id], 'b', linewidth=1.5)
        axes[0].axhline(self.J_th_T2_global, color='r', linestyle='--', linewidth=2)
        axes[0].set_title(f'T² Global MICA - IDV[{id}]')
        axes[0].set_xlabel('Amostras')
        axes[0].set_ylabel('T²')
        axes[0].grid(True)
        
        # Q Global
        axes[1].plot(Q_global[id], 'g', linewidth=1.5)
        axes[1].axhline(self.J_th_Q_global, color='r', linestyle='--', linewidth=2)
        axes[1].set_title(f'Q Global MICA - IDV[{id}]')
        axes[1].set_xlabel('Amostras')
        axes[1].set_ylabel('Q')
        axes[1].grid(True)
        
        # Q por Bloco
        for i in range(self.n_blocks):
            axes[2 + i].plot(Q_blocks[id][i], linewidth=1.5)
            axes[2 + i].axhline(self.J_th_Q_blocks[i], color='r', linestyle='--', linewidth=2)
            title = block_names[i] if i < len(block_names) else f'Bloco {i}'
            axes[2 + i].set_title(f'Q {title} MICA - IDV[{id}]')
            axes[2 + i].set_xlabel('Amostras')
            axes[2 + i].set_ylabel('Q')
            axes[2 + i].grid(True)
        
        plt.tight_layout()
        plt.show()
````

## 3.8. FDA (Fisher Discriminant Analysis)

### 3.8.1. Objetivo

O **FDA** (Análise Discriminante de Fisher) é um método de redução de dimensionalidade **supervisionado** que busca encontrar projeções lineares que **maximizem a separabilidade entre classes** (ou condições operacionais). Diferentemente de PCA (não supervisionado) e ICA (sem rótulos prévios), o FDA utiliza informação de classe para otimizar as direções de projeção.

No monitoramento de processos, o FDA é treinado com dados históricos etiquetados (ex: condições normais vs. falhas conhecidas) e aprende a encontrar direções que separem melhor essas condições, resultando em detecção mais precisa.

### 3.8.2. Formulação Matemática

Para um problema com $c$ classes, o FDA busca uma matriz de projeção **W** que maximize a razão:

$$J(\mathbf{W}) = \frac{|\mathbf{W}^\top \mathbf{S}_B \mathbf{W}|}{|\mathbf{W}^\top \mathbf{S}_W \mathbf{W}|}$$

Onde:
- **$\mathbf{S}_B$**: Matriz de covariância entre classes (between-class scatter):
  $$\mathbf{S}_B = \sum_{i=1}^{c} n_i (\mathbf{m}_i - \mathbf{m}) (\mathbf{m}_i - \mathbf{m})^\top$$

- **$\mathbf{S}_W$**: Matriz de covariância dentro das classes (within-class scatter):
  $$\mathbf{S}_W = \sum_{i=1}^{c} \sum_{\mathbf{x} \in C_i} (\mathbf{x} - \mathbf{m}_i) (\mathbf{x} - \mathbf{m}_i)^\top$$

- **$\mathbf{m}_i$**: Centroide da classe $i$
- **$\mathbf{m}$**: Centroide global

O FDA resolve o problema de autovalores generalizados:

$$\mathbf{S}_B \mathbf{w} = \lambda \mathbf{S}_W \mathbf{w}$$

Os autovetores associados aos maiores autovalores formam a matriz de projeção **W**.

### 3.8.3. Limites de Controle

Em um contexto de monitoramento (onde temos uma classe "normal" e potencialmente múltiplas classes de falha), o FDA projeta os dados de teste e calcula:

- **Distância de Mahalanobis** para medir proximidade à classe normal.
- **T²** de Hotelling nos escores LDA projetados.
- **Q** (SPE) do espaço residual não capturado.

Limites podem ser estabelecidos a partir da distribuição da distância no conjunto de treino normal.

### Aplicação no Processo Tennessee Eastman

Para este trabalho, treinamos o FDA com rótulos binários (operação normal vs. falha) utilizando dados de treino com falhas introduzidas. O modelo aprende a separar o espaço operacional normal do espaço de falhas, permitindo detecção de anomalias em tempo real mesmo sem conhecimento prévio do tipo específico de falha.

````python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

class FDAFaultDetector:
    def __init__(self, n_components=9, alpha=0.01):
        """
        Monitoramento de Falhas baseado em FDA (Fisher Discriminant Analysis).
        
        O treinamento requer dados rotulados (classe 0 = normal, classe 1 = falha).
        
        :param n_components: Número de componentes discriminantes.
        :param alpha: Nível de significância para limites de controle.
        """
        self.l = n_components
        self.alpha = alpha
        
        self.scaler = StandardScaler()
        # LDA do sklearn implementa FDA
        self.lda = LinearDiscriminantAnalysis(n_components=min(n_components, 1))  # LDA tem limite: n_classes-1
        
        # Parâmetros aprendidos
        self.var_scores = None
        self.J_th_T2 = None
        self.J_th_Q = None
        self.lda_mean_normal = None  # Centroide da classe normal
        self.lda_cov_normal = None   # Covariância da classe normal

    def fit(self, X_train, y_train):
        """
        Treina o FDA com dados rotulados.
        
        :param X_train: Dados de treino
        :param y_train: Rótulos (0 = normal, 1 = falha)
        """
        N = X_train.shape[0]
        
        # 1. Normalização
        Z_train = self.scaler.fit_transform(X_train)
        
        # 2. Treinamento LDA (FDA)
        self.lda.fit(Z_train, y_train)
        
        # 3. Projeção nos componentes LDA
        T_train = self.lda.transform(Z_train)
        self.var_scores = np.var(T_train, axis=0) if T_train.shape[1] > 1 else np.array([np.var(T_train)])
        
        # 4. Dados normais para calcular centroide e covariância
        mask_normal = y_train == 0
        Z_normal = Z_train[mask_normal]
        T_normal = T_train[mask_normal]
        
        self.lda_mean_normal = np.mean(T_normal, axis=0)
        
        cov_raw = np.cov(T_normal.T) if T_normal.shape[0] > 1 else np.eye(T_normal.shape[1])
        self.lda_cov_normal = np.atleast_2d(cov_raw)
        
        # 5. Cálculo de Resíduos
        # Reconstrução manual: X_rec = T * inversa(W) + médias
        W_pinv = np.linalg.pinv(self.lda.scalings_)
        X_rec = np.dot(T_train, W_pinv) + self.lda.xbar_
        E = Z_train - X_rec
        Q_train = np.sum(E**2, axis=1)
        
        # 6. Cálculo do T2 de Hotelling
        cov_inv = np.linalg.pinv(self.lda_cov_normal)
        diffs = T_train - self.lda_mean_normal
        T2_train = np.sum((diffs @ cov_inv) * diffs, axis=1)
        
        # 7. Limites
        # T2 Limit
        n_comp_eff = min(T_train.shape[1], self.l)
        F_alpha = f.ppf(1 - self.alpha, dfn=n_comp_eff, dfd=(N - n_comp_eff))
        self.J_th_T2 = ((n_comp_eff * (N**2 - 1)) / (N * (N - n_comp_eff))) * F_alpha
        
        # Q Limit (Box)
        m, v = np.mean(Q_train), np.var(Q_train)
        g, h = v / (2 * m), (2 * m**2) / v
        self.J_th_Q = g * chi2.ppf(1 - self.alpha, df=h)
        
        # DEBUG
        print(f"--- Treinamento FDA Concluído ---")
        print(f"Componentes LDA: {T_train.shape[1]} | Alpha: {self.alpha}")
        print(f"J_th_T2 (Limite T²): {self.J_th_T2:.4f}")
        print(f"J_th_Q  (Limite Q):  {self.J_th_Q:.4f}\n")

    def test_batch(self, X_test, y_test=None):
        """
        Testa o FDA em um novo lote de dados.
        
        :param X_test: Dados de teste
        :param y_test: (Opcional) Rótulos verdadeiros
        :return: T2, Q, y_pred
        """
        # 1. Normalização
        Z_test = self.scaler.transform(X_test)
        
        # 2. Projeção LDA
        T_test = self.lda.transform(Z_test)
        
        # 3. Cálculo do T2 (Distância de Mahalanobis)
        cov_inv = np.linalg.pinv(self.lda_cov_normal)
        diffs = T_test - self.lda_mean_normal
        T2 = np.sum((diffs @ cov_inv) * diffs, axis=1)
        
        # 4. Cálculo do Q
        # Reconstrução manual no teste
        W_pinv = np.linalg.pinv(self.lda.scalings_)
        X_rec = np.dot(T_test, W_pinv) + self.lda.xbar_
        E = Z_test - X_rec
        Q = np.sum(E**2, axis=1)
        
        # 5. Decisão de Falha
        y_pred = ((T2 > self.J_th_T2) | (Q > self.J_th_Q)).astype(int)
        
        return T2, Q, y_pred

    def plot_statistics(self, T2, Q, id=10):
        """
        Plota as estatísticas T2 e Q.
        """
        plt.figure(figsize=(16, 5))
        
        # Gráfico T2
        plt.subplot(1, 2, 1)
        plt.plot(T2[id], 'b', linewidth=2, label='T² (Mahalanobis)')
        plt.axhline(self.J_th_T2, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'T² FDA - IDV[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('T²')
        plt.legend()
        plt.grid()
        
        # Gráfico Q
        plt.subplot(1, 2, 2)
        plt.plot(Q[id], 'g', linewidth=2, label='Q (SPE)')
        plt.axhline(self.J_th_Q, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'Q FDA - IDV[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('Q')
        plt.legend()
        plt.grid()
        
        plt.tight_layout()
        plt.show()
````

## 3.9 Probabilistic Autoencoder (PAE)



### 3.9.1. Objetivo

O objetivo do Autoencoder Probabilístico (PAE) no monitoramento de processos é modelar o comportamento nominal (normal) do sistema de forma estritamente não-supervisionada, capturando as relações e correlações não-lineares complexas existentes entre as variáveis do processo. Ao mapear os dados operacionais para um espaço latente contínuo governado por distribuições de probabilidade, o PAE busca aprender a assinatura de estabilidade da planta para, posteriormente, detectar desvios e anomalias através de métricas de distorção latente e erro de reconstrução.

### 3.9.2. Formulação Matemática

O PAE (tipicamente implementado por meio de um *Variational Autoencoder* - VAE) é composto por duas redes neurais integradas que operam de forma probabilística:

1. **Codificador (*Encoder* - $q_\phi(z|x)$):** Em vez de mapear o dado de entrada $x$ para um ponto fixo, a rede com parâmetros determinísticos $\phi$ estima os parâmetros estatísticos (média $\mu_z$ e log-variância $\log \sigma_z^2$) de uma distribuição Gaussiana no espaço latente:

$$\mu_z = f_{\phi,\mu}(x), \quad \log \sigma_z^2 = f_{\phi,\sigma}(x)$$



Para viabilizar o treinamento via retropropagação, o vetor latente $z$ é amostrado utilizando o truque da reparametrização:

$$z = \mu_z + \sigma_z \odot \epsilon, \quad \text{onde } \epsilon \sim \mathcal{N}(0, I)$$


2. **Decodificador (*Decoder* - $p_\theta(x|z)$):** A rede com parâmetros $\theta$ recebe o vetor amostrado $z$ do espaço latente e tenta reconstruir as variáveis originais do processo no espaço de medição:

$$\hat{x} = g_\theta(z)$$


3. **Função de Perda ($\mathcal{L}_{PAE}$):** O modelo é otimizado minimizando o negativo do Limite Inferior de Evidência (ELBO), que equilibra a maximização da verossimilhança da reconstrução com a regularização do espaço latente via Divergência de Kullback-Leibler ($D_{KL}$):

$$\mathcal{L}_{PAE} = \text{Erro de Reconstrução} + D_{KL}(q_\phi(z|x) \parallel p(z))$$


$$\mathcal{L}_{PAE} = \|x - \hat{x}\|^2 - \frac{1}{2} \sum_{j=1}^{l} \left( 1 + \log\sigma_{zj}^2 - \mu_{zj}^2 - e^{\log\sigma_{zj}^2} \right)$$



### 3.9.3. Limites de Controle

Em um contexto de monitoramento não-supervisionado (onde o modelo é exposto apenas à dinâmica da classe normal durante o treino), o PAE projeta os novos dados de teste e calcula duas estatísticas complementares para a tomada de decisão:

* **Estatística $T^2$ Latente:** Avalia o comportamento das variáveis comprimidas dentro do espaço latente. Mede a distância de Mahalanobis dos escores médios ($\mu_z$) em relação à origem, penalizada pela variância dos escores obtida no treinamento ($\sigma_{z_{\text{treino}}}^2$):

$$T^2 = \sum_{j=1}^{l} \frac{\mu_{zj}^2}{\sigma_{z_{\text{treino}},j}^2}$$



O limite de controle crítico ($J_{th, T^2}$) é estabelecido teoricamente utilizando a distribuição $F$ de Snedecor sobre o conjunto de treino normal.
* **Estatística $Q$ (SPE - *Squared Prediction Error*):** Mede a magnitude dos resíduos de reconstrução no espaço original das variáveis. Representa a porção da variabilidade dos dados que o modelo não conseguiu decodificar:

$$Q = \sum_{i=1}^{m} (x_i - \hat{x}_i)^2$$



O limite de controle crítico ($J_{th, Q}$) é definido a partir da aproximação estatística de Box, utilizando a distribuição Qui-quadrado ($\chi^2$) calculada sobre a média e a variância dos resíduos de treino.

### Aplicação no Processo Tennessee Eastman

Para este trabalho, o PAE foi treinado de forma puramente não-supervisionada, utilizando exclusivamente a base de dados de operação normal (`df_CN_treino`). Ao contrário de abordagens como o FDA, o modelo dispensa o uso de rótulos ou exemplos de falhas na fase de ajuste de pesos. Ele especializa-se em compreender a dinâmica intrínseca e multidimensional da planta estável. Durante a etapa de testes online, a introdução de qualquer um dos distúrbios operacionais altera a correlação entre os sensores; consequentemente, o PAE deixa de reconstruir os dados de forma adequada (disparando o gráfico de resíduos $Q$) ou desloca as variáveis para fora da distribuição latente normal (disparando o gráfico $T^2$). Isso permite uma detecção rápida e robusta de anomalias em tempo real, sem a necessidade de conhecimento prévio ou mapeamento histórico da falha analisada.

````python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.preprocessing import StandardScaler
from scipy.stats import chi2, f
import matplotlib.pyplot as plt

# =========================================================================
# 1. Camada Customizada para calcular a perda (Loss) compatível com Keras 3
# =========================================================================
class VAELossLayer(layers.Layer):
    def call(self, inputs):
        encoder_inputs, outputs, z_mean, z_log_var = inputs
        
        # Dentro do método call, as operações nativas do TF rodam perfeitamente sem erros
        reconstruction_loss = tf.reduce_mean(tf.reduce_sum(tf.square(encoder_inputs - outputs), axis=1))
        kl_loss = -0.5 * tf.reduce_mean(tf.reduce_sum(1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var), axis=1))
        
        # Injeta a perda calculada diretamente no grafo do modelo
        self.add_loss(reconstruction_loss + kl_loss)
        return outputs

# =========================================================================
# 2. Classe Principal do Detector de Falhas Probabilístico
# =========================================================================
class PAEFaultDetector:
    def __init__(self, n_components=9, alpha=0.01, epochs=50, batch_size=32):
        """
        Monitoramento de Falhas baseado em Autoencoder Probabilístico (VAE).
        Compatível com Keras 3 e TensorFlow moderno.
        
        O treinamento é não-supervisionado (aprende apenas com dados da classe normal).
        """
        self.l = n_components
        self.alpha = alpha
        self.epochs = epochs
        self.batch_size = batch_size
        
        self.scaler = StandardScaler()
        
        # Parâmetros aprendidos
        self.var_scores = None
        self.J_th_T2 = None
        self.J_th_Q = None
        
        # Modelos Keras
        self.vae = None
        self.encoder = None

    def _build_model(self, input_dim):
        """Constrói a arquitetura do VAE ajustada para Keras 3."""
        # Encoder
        encoder_inputs = layers.Input(shape=(input_dim,))
        x = layers.Dense(32, activation='relu')(encoder_inputs)
        x = layers.Dense(16, activation='relu')(x)
        
        z_mean = layers.Dense(self.l, name='z_mean')(x)
        z_log_var = layers.Dense(self.l, name='z_log_var')(x)
        
        # Amostragem (Reparameterization trick)
        def sampling(args):
            z_mean, z_log_var = args
            batch = tf.shape(z_mean)[0]
            dim = tf.shape(z_mean)[1]
            epsilon = tf.random.normal(shape=(batch, dim))
            return z_mean + tf.exp(0.5 * z_log_var) * epsilon
            
        z = layers.Lambda(sampling, output_shape=(self.l,), name='z')([z_mean, z_log_var])
        self.encoder = models.Model(encoder_inputs, [z_mean, z_log_var, z], name='encoder')
        
        # Decoder
        latent_inputs = layers.Input(shape=(self.l,))
        x = layers.Dense(16, activation='relu')(latent_inputs)
        x = layers.Dense(32, activation='relu')(x)
        decoder_outputs = layers.Dense(input_dim, activation='linear')(x)
        
        self.decoder = models.Model(latent_inputs, decoder_outputs, name='decoder')
        
        # VAE Completo encadeado usando a camada customizada de perda (VAELossLayer)
        z_mean_enc, z_log_var_enc, z_enc = self.encoder(encoder_inputs)
        outputs_raw = self.decoder(z_enc)
        
        # Aplicação da camada que gerencia a perda internamente
        outputs = VAELossLayer()([encoder_inputs, outputs_raw, z_mean_enc, z_log_var_enc])
        
        self.vae = models.Model(encoder_inputs, outputs, name='vae')
        self.vae.compile(optimizer='adam')

    def fit(self, X_train, y_train=None):
        """Treina o PAE apenas com dados normais."""
        if y_train is not None:
            mask_normal = y_train == 0
            X_train = X_train[mask_normal]
            
        N = X_train.shape[0]
        input_dim = X_train.shape[1]
        
        # 1. Normalização
        Z_train = self.scaler.fit_transform(X_train)
        
        # 2. Construção e Treino
        self._build_model(input_dim)
        print("Treinando o Autoencoder Probabilístico (PAE)...")
        self.vae.fit(Z_train, Z_train, epochs=self.epochs, batch_size=self.batch_size, verbose=0)
        
        # 3. Projeção no Espaço Latente e Reconstrução (Retorna arrays NumPy puros)
        z_mean, _, _ = self.encoder.predict(Z_train, verbose=0)
        X_rec = self.vae.predict(Z_train, verbose=0)
        
        # 4. Cálculo do T2 (Hotelling latente)
        self.var_scores = np.var(z_mean, axis=0) + 1e-9
        T2_train = np.sum((z_mean**2) / self.var_scores, axis=1)
        
        # 5. Cálculo do Q (Resíduos de Reconstrução)
        E = Z_train - X_rec
        Q_train = np.sum(E**2, axis=1)
        
        # 6. Limites de Controle
        # T2 Limit (F-distribution)
        n_comp_eff = min(z_mean.shape[1], self.l)
        F_alpha = f.ppf(1 - self.alpha, dfn=n_comp_eff, dfd=(N - n_comp_eff))
        self.J_th_T2 = ((n_comp_eff * (N**2 - 1)) / (N * (N - n_comp_eff))) * F_alpha
        
        # Q Limit (Box)
        m, v = np.mean(Q_train), np.var(Q_train)
        g, h = v / (2 * m), (2 * m**2) / v
        self.J_th_Q = g * chi2.ppf(1 - self.alpha, df=h)
        
        # DEBUG
        print(f"--- Treinamento PAE Concluído ---")
        print(f"Componentes Latentes: {z_mean.shape[1]} | Alpha: {self.alpha}")
        print(f"J_th_T2 (Limite T²): {self.J_th_T2:.4f}")
        print(f"J_th_Q  (Limite Q):  {self.J_th_Q:.4f}\n")

    def test_batch(self, X_test, y_test=None):
        """Testa o PAE em um novo lote de dados de teste."""
        # 1. Normalização
        Z_test = self.scaler.transform(X_test)
        
        # 2. Projeção Latente
        z_mean, _, _ = self.encoder.predict(Z_test, verbose=0)
        
        # 3. Cálculo do T2
        T2 = np.sum((z_mean**2) / self.var_scores, axis=1)
        
        # 4. Cálculo do Q
        X_rec = self.vae.predict(Z_test, verbose=0)
        E = Z_test - X_rec
        Q = np.sum(E**2, axis=1)
        
        # 5. Decisão de Falha
        y_pred = ((T2 > self.J_th_T2) | (Q > self.J_th_Q)).astype(int)
        
        return T2, Q, y_pred

    def plot_statistics(self, T2, Q, id=10):
        """Plota as estatísticas T2 e Q coletadas."""
        plt.figure(figsize=(16, 5))
        
        # Gráfico T2
        plt.subplot(1, 2, 1)
        plt.plot(T2[id], 'b', linewidth=2, label='T² Latente')
        plt.axhline(self.J_th_T2, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'T² PAE - IDV[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('T²')
        plt.legend()
        plt.grid()
        
        # Gráfico Q
        plt.subplot(1, 2, 2)
        plt.plot(Q[id], 'g', linewidth=2, label='Q (SPE)')
        plt.axhline(self.J_th_Q, color='r', linestyle='--', linewidth=2, label='Limiar')
        plt.title(f'Q PAE - IDV[{id}]')
        plt.xlabel('Amostras')
        plt.ylabel('Q')
        plt.legend()
        plt.grid()
        
        plt.tight_layout()
        plt.show()
````

````python
# =============================================================
# Preparação de Dados para PAE
# =============================================================
# Configuração da Monte Carlo
n_mc = 10  # número de execuções independentes

# Listas para armazenar os resultados de cada rodada
FAR_mc = []
FDRs_mc = []  # será uma lista de listas: FDRs_mc[rodada][falha]

# Vamos usar APENAS os dados normais (df_CN_treino).
colunas_pae = [f"XMEAS({i})" for i in range(1, 23)] + [f"XMV({i})" for i in range(1, 12)]
X_pae_train = df_CN_treino[colunas_pae].values

# Rótulos verdadeiros para os dados de teste (fixos)
# Primeiro lote: apenas dados normais
y_true_normal = np.zeros(dados_teste[0].shape[0])
# Lotes de falha: 160 amostras normais + 800 com falha (exemplo do seu código)
y_true_commun = np.concatenate([np.zeros(160), np.ones(800)])

y_test_true_for_fault_pae = [y_true_normal]
for i in range(len(dados_teste) - 1):
    y_test_true_for_fault_pae.append(y_true_commun)

# =============================================================
# Loop de Monte Carlo
# =============================================================
print(f"Iniciando Monte Carlo com {n_mc} repetições...\n")

for run in range(1,n_mc+1):
    print(f"--- Execução {run}/{n_mc} ---")

     # 1. Criar e treinar o modelo (nova inicialização aleatória)
    pae_monitor = PAEFaultDetector(n_components=9, alpha=0.01, epochs=50)
    pae_monitor.fit(X_pae_train)

    # 2. Avaliar em todos os lotes de teste
    T2_for_fault_pae = []
    Q_for_fault_pae = []
    y_test_pred_for_fault_pae = []

    for df_test in dados_teste:
        X_test_batch = df_test[colunas_pae].values
        T2, Q, y_pred = pae_monitor.test_batch(X_test_batch)
        T2_for_fault_pae.append(T2)
        Q_for_fault_pae.append(Q)
        y_test_pred_for_fault_pae.append(y_pred)

    # 3. Calcular FAR (lote normal) e FDRs (lotes de falha)
    FAR_val = (np.sum(y_test_pred_for_fault_pae[0]) / len(y_test_true_for_fault_pae[0])) * 100
    FAR_mc.append(FAR_val)
    
    # FDR para cada falha (índices 1 em diante)
    FDRs_run = []
    for i, (y_true, y_pred) in enumerate(zip(y_test_true_for_fault_pae[1:], 
                                             y_test_pred_for_fault_pae[1:]), start=1):
        # FDR = acertos nos instantes com falha (y_true == 1)
        fdr_val = (np.sum(y_pred[y_true == 1]) / np.sum(y_true[y_true == 1])) * 100
        FDRs_run.append(fdr_val)
    
    FDRs_mc.append(FDRs_run)
    
    # (Opcional) Mostrar métricas da rodada atual
    print(f"  FAR = {FAR_val:.2f}%")
    #for idx, fdr in enumerate(FDRs_run, start=1):
    #    print(f"  FDR IDV({idx:02d}) = {fdr:.2f}%")
    #print()

# =============================================================
# Cálculo das médias e desvios padrão
# =============================================================
FAR_mean = np.mean(FAR_mc)
FAR_std  = np.std(FAR_mc, ddof=1)   # desvio padrão amostral

print("=" * 50)
print(f"RESULTADOS FINAIS (média ± desvio padrão sobre {n_mc} rodadas)")
print("-" * 50)
print(f"Taxa de Alarme Falso (FAR) - PAE: {FAR_mean:.2f}% ± {FAR_std:.2f}%")


# Converter FDRs_mc para array (rodadas x falhas) e calcular média/desvio por falha
FDRs_array = np.array(FDRs_mc)  # shape (n_mc, num_falhas)
num_falhas = FDRs_array.shape[1]
print("-" * 50)
for i in range(num_falhas):
    falha_id = i + 1
    fdr_mean = np.mean(FDRs_array[:, i])
    fdr_std  = np.std(FDRs_array[:, i], ddof=1)
    print(f"FDR para a Falha IDV({falha_id:02d}) - PAE: {fdr_mean:.2f}% ± {fdr_std:.2f}%")

print("=" * 50)

# Média do FAR
FAR_pae = FAR_mean  

# Média dos FDRs (uma lista com a média para cada falha)
FDRs_pae = [np.mean(FDRs_array[:, i]) for i in range(FDRs_array.shape[1])]
````
